# Building a Chatbot Using LangGraph

# 1. What Are We Going to Build?

We are going to build a **basic conversational chatbot using LangGraph**.

The chatbot will:

1. Accept a user message.
2. Store the conversation state.
3. Send the conversation to an LLM.
4. Generate an AI response.
5. Store the response in state.
6. Continue the conversation when the user sends another message.

### Basic Architecture

    User
      ↓
    User Message
      ↓
    LangGraph State
      ↓
    Chatbot Node
      ↓
    LLM
      ↓
    AI Response
      ↓
    Update State
      ↓
    Response to User


---

# 2. What Is a LangGraph Chatbot?

A LangGraph chatbot is a conversational application where **LangGraph manages the workflow and conversation state**, while an LLM generates responses.

### Simple Definition

> A LangGraph chatbot is an LLM-powered conversational system whose conversation flow and state are managed using a LangGraph graph.

The simplest architecture is:

    START
      ↓
    Chatbot Node
      ↓
    END


But internally, the chatbot node interacts with the LLM and conversation messages.


---

# 3. Why Use LangGraph for a Chatbot?

You can build a simple chatbot directly using an LLM API.

So why use LangGraph?

Because LangGraph becomes valuable when the chatbot needs:

- Conversation state
- Multiple workflow steps
- Memory
- Tools
- Conditional routing
- Loops
- Human-in-the-loop
- RAG
- Database access
- Agentic behavior
- Multiple agents
- Persistence
- Error handling

A simple chatbot:

    User → LLM → Response

A LangGraph chatbot can evolve into:

                         ┌→ Search Tool ──┐
                         │                │
    User → Chatbot → Router              ├→ Response
                         │                │
                         └→ Database ────┘

# 11. Chatbot State

State is one of the most important concepts.

A chatbot needs to remember the conversation.

Conceptually:

    class State(TypedDict):
        messages: list


The state contains:

    messages


Example:

    {
        "messages": [
            HumanMessage("Hello"),
            AIMessage("Hi! How can I help?")
        ]
    }


After another user message:

    {
        "messages": [
            HumanMessage("Hello"),
            AIMessage("Hi! How can I help?"),
            HumanMessage("What is LangGraph?")
        ]
    }


After the LLM responds:

    {
        "messages": [
            HumanMessage("Hello"),
            AIMessage("Hi! How can I help?"),
            HumanMessage("What is LangGraph?"),
            AIMessage("LangGraph is...")
        ]
    }

# 12. Why Messages Instead of Just Strings?

A conversation contains different types of messages.

### Human Message

    HumanMessage

Represents the user.

### AI Message

    AIMessage

Represents the assistant.

### System Message

    SystemMessage

Defines instructions or behavior.

### Tool Message

    ToolMessage

Contains the result of a tool call.


Example:

    System
    ↓
    Human
    ↓
    AI
    ↓
    Human
    ↓
    AI


Using structured messages allows the LLM to understand the role of each message.

# 14. Message Reducer

There is an important problem.

Suppose the state contains:

    messages = [old messages]

The chatbot produces:

    new AI message

We don't want the new message to completely replace the old messages.

We want:

    old messages
         +
    new message

This is where a **message reducer** becomes important.

LangGraph provides message-oriented state patterns that allow new messages to be appended rather than simply replacing the existing list.


---

# 15. Modern Message State

For chat applications, LangGraph provides convenient message-state abstractions.

Conceptually:

    from langgraph.graph import MessagesState


    class State(MessagesState):
        pass


This gives the graph a state structure centered around messages.

Another common pattern is to use:

    messages

with an appropriate reducer such as:

    add_messages


The important concept is:

> New messages are merged into the conversation history instead of replacing previous messages.

# 16. Basic Chatbot Node

The chatbot node receives state:

    def chatbot(state):
        response = llm.invoke(state["messages"])

        return {
            "messages": [response]
        }


Flow:

    State
      ↓
    chatbot()
      ↓
    state["messages"]
      ↓
    LLM
      ↓
    AI Response
      ↓
    Return new message


The node's responsibility is simple:

> Send the conversation to the LLM and return the generated AI message.

# 17. Complete Basic Chatbot Code

Here is the simplest practical structure.

    from langgraph.graph import StateGraph, START, END, MessagesState
    from langchain_openai import ChatOpenAI


    llm = ChatOpenAI(
        model="YOUR_MODEL_NAME"
    )


    class State(MessagesState):
        pass


    def chatbot(state: State):
        response = llm.invoke(state["messages"])

        return {
            "messages": [response]
        }


    graph = StateGraph(State)

    graph.add_node("chatbot", chatbot)

    graph.add_edge(START, "chatbot")
    graph.add_edge("chatbot", END)

    app = graph.compile()


    result = app.invoke({
        "messages": [
            {
                "role": "user",
                "content": "Hello! My name is Musharraf."
            }
        ]
    })


    print(result["messages"][-1].content)


The exact model name and API setup depend on the LLM provider you choose.

# 30. But There Is a Problem

The basic example:

    app.invoke(...)

runs one graph execution.

If we call it again with only:

    "What is LangGraph?"

the previous conversation may not automatically be available unless we provide/persist it appropriately.

We need **conversation persistence** for a real chatbot.

This leads to the next major concept:

    Checkpointing


---

# 31. Chatbot Memory

There are two related concepts:

### State

Information available during the current graph execution.

### Persistence / Checkpointing

Saving graph state so it can be recovered across separate invocations.

For a chatbot:

    User Message
         ↓
    State
         ↓
    LLM
         ↓
    Updated State
         ↓
    Checkpoint
         ↓
    Next Message
         ↓
    Restore State
         ↓
    LLM

# 32. Thread ID

A conversation needs an identity.

For example:

    thread_id = "user_123_conversation_1"


This allows LangGraph to distinguish:

    Conversation A

from:

    Conversation B


Architecture:

    Thread A
    ├── Hello
    ├── What is LangGraph?
    └── Explain nodes

    Thread B
    ├── Hello
    ├── What is Python?
    └── Explain decorators


Each thread has its own conversation state.


---

# 33. In-Memory Checkpointing

For learning, we can use an in-memory checkpointer.

Conceptually:

    from langgraph.checkpoint.memory import InMemorySaver

    checkpointer = InMemorySaver()

    app = graph.compile(
        checkpointer=checkpointer
    )


Now the graph can persist state during the application's lifetime.

# 37. Complete Basic Memory Chatbot

Conceptual code:

    from langgraph.graph import StateGraph, START, MessagesState
    from langgraph.checkpoint.memory import InMemorySaver
    from langchain_openai import ChatOpenAI


    llm = ChatOpenAI(
        model="YOUR_MODEL_NAME"
    )


    class State(MessagesState):
        pass


    def chatbot(state: State):
        response = llm.invoke(state["messages"])

        return {
            "messages": [response]
        }


    graph = StateGraph(State)

    graph.add_node("chatbot", chatbot)

    graph.add_edge(START, "chatbot")

    checkpointer = InMemorySaver()

    app = graph.compile(
        checkpointer=checkpointer
    )


    config = {
        "configurable": {
            "thread_id": "conversation_1"
        }
    }


    response_1 = app.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": "My name is Musharraf."
                }
            ]
        },
        config
    )


    print(response_1["messages"][-1].content)


    response_2 = app.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": "What is my name?"
                }
            ]
        },
        config
    )


    print(response_2["messages"][-1].content)


The important point is that both calls use:

    thread_id = "conversation_1"


Therefore they belong to the same conversation.

# 57. Interview Questions

### Q1. Why use LangGraph to build a chatbot?

To manage stateful, multi-step, controllable conversational workflows and provide a foundation for tools, persistence, conditional routing, and agentic behavior.

### Q2. What is the role of the LLM?

The LLM understands the conversation and generates responses.

### Q3. What is the role of LangGraph?

LangGraph orchestrates the workflow and manages state transitions.

### Q4. What is chatbot state?

The information maintained by the graph during execution, commonly including conversation messages.

### Q5. What is MessagesState?

A LangGraph state abstraction designed around message-based conversation state.

### Q6. Why are messages important?

They preserve the roles and structure of the conversation, such as human, AI, system, and tool messages.

### Q7. What is a checkpointer?

A mechanism used to save and restore graph state across executions.

### Q8. What is a thread ID?

An identifier used to associate graph executions with a particular conversation/state lineage.

### Q9. How can a chatbot become agentic?

By adding tools, decision-making, conditional routing, observation, and iterative loops.

### Q10. What is the simplest LangGraph chatbot graph?

    START → Chatbot Node → END
