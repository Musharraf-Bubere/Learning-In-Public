# Streamlit — Detailed Notes for AI/ML & GenAI Applications

# 1. What is Streamlit?

**Streamlit** is an open-source Python framework used to build interactive web applications quickly, especially for:

- Data Science
- Machine Learning
- Generative AI
- LLM applications
- AI Agents
- Data dashboards
- Model demonstrations
- Prototypes

### Simple Definition

> Streamlit allows us to convert Python code into an interactive web application without requiring traditional frontend technologies such as HTML, CSS, and JavaScript.

### Traditional Web Application

    Frontend
    HTML
    CSS
    JavaScript
       ↓
    Backend
    Python / FastAPI
       ↓
    Model / Database


### Streamlit Application

    Python Code
       ↓
    Streamlit
       ↓
    Web Application


This makes Streamlit particularly useful for AI/ML developers.

# 4. Streamlit vs Traditional Web Development

| Feature | Streamlit | Traditional Web |
|---|---|---|
| Main language | Python | Frontend + Backend |
| HTML required | Usually no | Yes |
| CSS required | Usually no | Usually yes |
| JavaScript required | Usually no | Usually yes |
| Development speed | Very fast | More involved |
| ML integration | Excellent | Requires API integration |
| Prototyping | Excellent | Slower |
| Production flexibility | Moderate | High |
| Custom UI control | Moderate | Very high |
| AI demos | Excellent | More work |

### Important

Streamlit is not necessarily a replacement for every frontend framework.

It is especially powerful when:

> Python + Data + AI + Rapid UI development

are the main requirements.

# 7. How Streamlit Executes Code

One important concept:

> Streamlit normally reruns the Python script when the user interacts with widgets.

Example:

    User changes input
          ↓
    Streamlit reruns script
          ↓
    UI updates


This is different from the traditional:

    Browser → API request → specific backend function


mental model.

# 40. Session State

One of the MOST IMPORTANT Streamlit concepts.

### Problem

Streamlit reruns the script when the user interacts with the application.

We need a way to preserve values across reruns.

That is where:

    st.session_state

comes in.


---

# 41. What is Session State?

> `st.session_state` is Streamlit's mechanism for storing values across reruns for a user's current browser session.

Example:

    if "count" not in st.session_state:
        st.session_state.count = 0


    if st.button("Increment"):
        st.session_state.count += 1


    st.write(
        st.session_state.count
    )


Without session state, values may be reset during reruns.


---

# 42. Session State Mental Model

    User Session
         ↓
    st.session_state
         ↓
    ┌───────────────┐
    │ count = 5     │
    │ messages = [] │
    │ user_id = ... │
    └───────────────┘
         ↓
    Script Reruns
         ↓
    State remains available


Important:

> Streamlit session state is application/UI session state. It is not the same thing as LangGraph persistence.


---

### Streamlit Session State

### LangGraph Persistence

Used for:

Architecture:

    Streamlit

    session_state

They can work together.

# 47. Basic Streamlit Chatbot

Simple example:

    import streamlit as st

    st.title("Simple Chatbot")

    if "messages" not in st.session_state:
        st.session_state.messages = []


    for message in st.session_state.messages:

        with st.chat_message(
            message["role"]
        ):
            st.write(message["content"])


    prompt = st.chat_input(
        "Ask something..."
    )


    if prompt:

        st.session_state.messages.append({
            "role": "user",
            "content": prompt
        })

        with st.chat_message("user"):
            st.write(prompt)

        response = f"You said: {prompt}"

        st.session_state.messages.append({
            "role": "assistant",
            "content": response
        })

        with st.chat_message("assistant"):
            st.write(response)


This is a basic UI-only chatbot.

# 50. Basic Streamlit + LangGraph Code

Conceptual example:

    import streamlit as st

    from langgraph.graph import (
        StateGraph,
        START,
        MessagesState
    )

    from langchain_openai import ChatOpenAI


    llm = ChatOpenAI(
        model="YOUR_MODEL_NAME"
    )


    class State(MessagesState):
        pass


    def chatbot(state: State):

        response = llm.invoke(
            state["messages"]
        )

        return {
            "messages": [response]
        }


    graph = StateGraph(State)

    graph.add_node(
        "chatbot",
        chatbot
    )

    graph.add_edge(
        START,
        "chatbot"
    )


    app = graph.compile()


    st.title("LangGraph Chatbot")


    if "messages" not in st.session_state:
        st.session_state.messages = []


    for message in st.session_state.messages:

        with st.chat_message(
            message["role"]
        ):
            st.write(
                message["content"]
            )


    prompt = st.chat_input(
        "Ask something..."
    )


    if prompt:

        st.session_state.messages.append({
            "role": "user",
            "content": prompt
        })


        with st.chat_message("user"):
            st.write(prompt)


        result = app.invoke({
            "messages": [
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        })


        response = result[
            "messages"
        ][-1].content


        st.session_state.messages.append({
            "role": "assistant",
            "content": response
        })


        with st.chat_message("assistant"):
            st.write(response)


Important:

This example demonstrates the UI + graph connection, but it does not yet provide durable conversation persistence through a LangGraph checkpointer.

    User
      ↓
    LLM
      ↓

      ↓
    Complete response
      ↓

    User
      ↓

      ↓
    "Hello, how"
      ↓

# 55. Why Streaming?

Without streaming:

    User
      ↓
    Wait 5 seconds
      ↓
    Full response


With streaming:

    User
      ↓
    First token appears
      ↓
    More tokens
      ↓
    More tokens
      ↓
    Complete response


Benefits:

- Better user experience
- Lower perceived latency
- Feels more responsive
- Natural for chat applications

# 57. Simple Streaming Example

    import time
    import streamlit as st


    def response_generator():

        for word in [
            "Hello ",
            "this ",
            "is ",
            "a ",
            "streaming ",
            "response."
        ]:

            time.sleep(0.3)

            yield word

    if st.button("Generate"):

            response_generator()

# 61. Types of Streaming to Understand

When working with LangGraph + Streamlit, distinguish between:

### 1. Token/content streaming

Stream generated LLM content.

    "Hel"
    "Hello"
    "Hello world"


### 2. State/update streaming

Stream graph state updates as nodes execute.

    Node A completed
       ↓
    State update
       ↓
    Node B completed


### 3. Message streaming

Stream messages/chunks generated by models or tools.

### 4. Custom event/progress streaming

Stream application-specific events such as:

    "Searching documents..."
    "Running SQL..."
    "Generating report..."


The exact streaming modes available depend on the LangGraph/LangChain versions and APIs being used.

If the same inputs are used, Streamlit can reuse the cached result.

# 69. st.cache_resource

Use for resources that should be reused.

Examples:

    ML model
    LLM client
    Database connection
    Vector store


Conceptually:

    @st.cache_resource
    def load_model():
        return model


This prevents expensive resources from being recreated unnecessarily.

# 70. Cache Data vs Cache Resource

| Feature | cache_data | cache_resource |
|---|---|---|
| Purpose | Computed/data results | Shared resources |
| Example | DataFrame | ML model |
| Example | API result | DB connection |
| Typical use | Data processing | Resource initialization |

### Easy memory trick

> `cache_data` → Cache the result.

> `cache_resource` → Cache the thing that produces results.

# 109. Common Mistakes

### Mistake 1

Thinking Streamlit is a normal frontend framework.

It is a Python-first application framework with its own execution model.


### Mistake 2

Ignoring reruns.

Streamlit reruns the script frequently.


### Mistake 3

Not using session state for UI state.

Important values can otherwise disappear across reruns.


### Mistake 4

Using session state as a production database.

It is not a replacement for durable persistence.


### Mistake 5

Creating the LLM/model repeatedly.

Use appropriate resource caching.


### Mistake 6

Blocking the UI with long operations.

Use streaming, status indicators, or appropriate async patterns.


### Mistake 7

Duplicating LangGraph state and Streamlit state unnecessarily.

Define a clear source of truth.


### Mistake 8

Putting all code into `app.py`.

Separate UI, graph, nodes, tools, and configuration as the project grows.

# 110. Streamlit Interview Questions

### Q1. What is Streamlit?

A Python framework for building interactive web applications, especially for data and AI applications.

### Q2. Why is Streamlit popular in Data Science?

It allows Python developers to quickly create interactive applications without building a traditional frontend.


### Q3. What causes a Streamlit rerun?

User interactions with widgets and certain application events can cause the script to rerun.


### Q4. What is `st.session_state`?

A mechanism for storing values across reruns within a user's Streamlit session.


### Q5. What is `st.cache_data`?

A caching mechanism for data/computation results.


### Q6. What is `st.cache_resource`?

A caching mechanism for reusable resources such as models or connections.


### Q7. What are `st.chat_input()` and `st.chat_message()`?

They are Streamlit components specifically designed for chat interfaces.


### Q8. How can you stream an LLM response in Streamlit?

Use a streaming generator/model stream and display it incrementally, commonly with `st.write_stream()`.


### Q9. What is the difference between Streamlit streaming and LangGraph streaming?

Streamlit provides the UI mechanism for displaying incremental output, while LangGraph provides mechanisms for streaming model messages, state updates, or workflow execution events.


### Q10. Can Streamlit be used with LangGraph?

Yes. Streamlit can serve as the UI while LangGraph manages stateful workflow orchestration.


### Q11. Can Streamlit be used with FastAPI?

Yes. They can be separate UI and API layers in a larger architecture.


### Q12. Is Streamlit suitable for production?

It can be used for production applications, especially internal tools and smaller AI applications, but production requirements such as authentication, scaling, observability, persistence, and security must still be addressed.
