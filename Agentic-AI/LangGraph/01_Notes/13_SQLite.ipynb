# SQLite — Complete Revision Notes

## 1. What is SQLite?

SQLite is a **lightweight, serverless, self-contained, file-based relational database management system (RDBMS)**.

Unlike MySQL or PostgreSQL, SQLite does not require a separate database server.

The entire database is usually stored in a **single `.db` / `.sqlite` file**.

### Simple Definition

> SQLite is an embedded relational database that stores structured data directly in a local database file.

### Example

A database can simply be:

    app.db

Inside this file we can have:

    users
    products
    orders
    messages

---

# 2. Why SQLite?

SQLite is useful when an application needs a database but does not need a separate database server.

### Common use cases

- Local applications
- Desktop applications
- Mobile applications
- Prototyping
- Testing
- Small applications
- Embedded systems
- Data analysis
- Learning SQL
- Development environments
- AI/ML prototypes
- Small FastAPI/Flask applications

### Example

For a small Python application:

    Python Application
          |
          v
       SQLite
          |
          v
       app.db

No database server is required.

# 3. SQLite vs MySQL/PostgreSQL

| Feature | SQLite | MySQL | PostgreSQL |
|---|---|---|---|
| Architecture | Embedded | Client-server | Client-server |
| Server required | No | Yes | Yes |
| Database storage | File | Server-managed | Server-managed |
| Setup | Very easy | Moderate | Moderate |
| Best for | Small/local apps | Web applications | Large/complex systems |
| Concurrent writes | Limited | Better | Excellent |
| SQL support | Good | Extensive | Very extensive |
| Transactions | Yes | Yes | Yes |
| Foreign keys | Supported | Supported | Supported |
| Scaling | Limited | High | Very high |

### Shortcut

    SQLite       → Local/simple
    MySQL        → Web/general-purpose
    PostgreSQL   → Complex/production systems

# 5. Important Characteristics of SQLite

## 5.1 Serverless

There is no separate database server process.

## 5.2 Zero Configuration

No complicated installation or server configuration is required.

## 5.3 Self-contained

SQLite contains the database engine and stores the database in a file.

## 5.4 Transactional

# 9. SQLite Command-Line Interface

After installing SQLite CLI, we can create/open a database:

    sqlite3 app.db

Useful commands:

    .databases

Shows databases.

    .tables

Shows tables.

    .schema users

Shows the schema of the users table.

    .headers on

Displays column names.

    .mode column

Displays results in column format.

    .quit

Exit SQLite.

Important:

Commands beginning with `.` are SQLite shell commands, not SQL commands.

Example:

    .tables

is a SQLite CLI command.

Whereas:

    SELECT * FROM users;

is SQL.

# 10. Creating a Database

SQLite automatically creates a database file when needed.

Example:

    sqlite3 app.db

If `app.db` does not exist, SQLite creates it.

In Python:

    import sqlite3

    connection = sqlite3.connect("app.db")

This creates the database file if it doesn't already exist.

# 11. Connection

A connection represents the communication between the Python application and SQLite database.

Example:

    import sqlite3

    connection = sqlite3.connect("app.db")

Now Python can execute SQL commands using this connection.

---

# 12. Cursor

A cursor is used to execute SQL statements and retrieve results.

Example:

    cursor = connection.cursor()

Then:

    cursor.execute("SELECT * FROM users")

The cursor can fetch the result.

# 13. Basic Python SQLite Workflow

The general workflow is:

    Import sqlite3
          ↓
    Connect to database
          ↓
    Create cursor
          ↓
    Execute SQL
          ↓
    Commit changes
          ↓
    Fetch results
          ↓

    connection = sqlite3.connect("app.db")

    cursor = connection.cursor()

    cursor.execute("SELECT * FROM users")

# 16. SQLite Type Affinity

SQLite does not enforce data types as strictly as some traditional RDBMS systems.

SQLite uses **type affinity**.

Common affinities:

- INTEGER
- TEXT
- REAL
- BLOB
- NUMERIC

Example:

    CREATE TABLE users (
        id INTEGER,
        name TEXT,
        age INTEGER
    );

SQLite tries to store values according to the column's affinity.

### Important

SQLite's flexible typing is convenient, but applications should still validate data properly.

# 17. PRIMARY KEY

A primary key uniquely identifies a row.

Example:

    CREATE TABLE users (
        id INTEGER PRIMARY KEY,
        name TEXT
    );

Each user gets a unique `id`.

For an `INTEGER PRIMARY KEY`, SQLite provides special behavior where the column is tied to the row identifier.

---

# 18. AUTOINCREMENT

Example:

    CREATE TABLE users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT
    );

However, `AUTOINCREMENT` is usually not necessary.

This is often sufficient:

    id INTEGER PRIMARY KEY

SQLite can automatically assign row IDs.

### Difference

    INTEGER PRIMARY KEY

Usually preferred.

    INTEGER PRIMARY KEY AUTOINCREMENT

Explicitly prevents reuse of certain previously generated ROWIDs, but has additional overhead.

# 19. INSERT

# 20. Parameterized Queries

Never build SQL using string concatenation when user input is involved.

Bad:

    name = "Ali"

    cursor.execute(
        f"SELECT * FROM users WHERE name = '{name}'"
    )

Better:

    cursor.execute(
        "SELECT * FROM users WHERE name = ?",
        (name,)
    )

Parameterized queries help protect against SQL injection.

# 23. UPDATE

Example:

    UPDATE users
    SET age = 26

    cursor.execute("""
        UPDATE users

# 24. DELETE

Example:

# 25. Transactions

A transaction is a group of database operations treated as one logical unit.

Example:

    BEGIN
       Operation 1
       Operation 2
       Operation 3
    COMMIT

If something fails:

    ROLLBACK

---

# 26. COMMIT

`commit()` permanently saves changes to the database.

Example:

    connection.commit()

Without committing appropriate write operations, changes may not be persisted as expected.

---

# 27. ROLLBACK

Rollback cancels uncommitted changes.

Example:

    try:
        cursor.execute(...)
        cursor.execute(...)

        connection.commit()

    except Exception:
        connection.rollback()

---

# 28. ACID Properties

SQLite supports transactional behavior based on ACID principles.

## Atomicity

All operations in a transaction happen or none do.

## Consistency

The database moves from one valid state to another.

## Isolation

Transactions are isolated according to SQLite's concurrency model.

## Durability

Committed data is designed to survive normal application/system failures.

### Shortcut

    A → All or nothing
    C → Valid state
    I → Transactions don't improperly interfere
    D → Committed data persists

---

# 29. Context Manager

Python provides a cleaner way to manage transactions.

Example:

    import sqlite3

    connection = sqlite3.connect("app.db")

    with connection:
        connection.execute(
            "INSERT INTO users (name) VALUES (?)",
            ("Ali",)
        )

The context manager handles transaction commit/rollback behavior based on whether an exception occurs.

# 30. Foreign Keys

Example:

    CREATE TABLE users (
        id INTEGER PRIMARY KEY,
        name TEXT

    CREATE TABLE orders (
        id INTEGER PRIMARY KEY,
        user_id INTEGER,

        FOREIGN KEY (user_id)

    users
      |

Python:

    connection.execute("PRAGMA foreign_keys = ON")

Then SQLite can enforce foreign-key constraints.

# 37. Indexes

An index improves lookup performance for appropriate queries.

Example:

    CREATE INDEX idx_users_email
    ON users(email);

Then searching:

    SELECT *
    FROM users

### Important

Indexes improve reads but have costs:

- Extra storage
- Additional write/update overhead
- Maintenance overhead

Don't create indexes on every column blindly.

# 38. Composite Index

Example:

# 39. Views

Example:

    SELECT * FROM adult_users;

A view normally does not store a separate copy of the result data.

# 52. SQLite Transactions and Concurrency

SQLite supports multiple readers and controlled concurrent access, but writing is more constrained than in many client-server databases.

Important idea:

    Many readers
         +
    Limited concurrent writing
         ↓
    SQLite concurrency model

SQLite is excellent for many local/small workloads but may not be the best choice for a heavily concurrent write-intensive application.

---

# 53. WAL Mode

SQLite supports **Write-Ahead Logging (WAL)**.

Enable it with:

    PRAGMA journal_mode = WAL;

WAL can improve concurrency between readers and writers for suitable workloads.

Conceptually:

Traditional journaling:

    Database
       ↑
    Journal

WAL:

    Readers → Database
    Writer  → WAL
                |
                v
             Database

WAL is especially useful in applications with simultaneous reads and writes.

---

# 54. SQLite Locking

SQLite uses locking mechanisms to coordinate access to the database.

A simplified idea:

    Reader → reads database
    Writer → obtains required write access
    Other conflicting operations → wait/fail according to configuration

For high-concurrency applications with many simultaneous writers, a client-server database may be a better choice.

---

# 55. Busy Timeout

When database access is temporarily blocked, applications can configure a timeout.

Python:

    connection = sqlite3.connect(
        "app.db",
        timeout=10
    )

This tells the connection how long to wait for a locked database before raising an error.

---

# 56. Row Factory

By default, Python's SQLite module often returns rows as tuples.

Example:

    (1, "Ali", "ali@example.com")

You can configure a row factory.

Example:

    connection.row_factory = sqlite3.Row

Then:

    row["name"]

can be used to access a column by name.

This is very useful in APIs and applications.

# 57. Complete Python CRUD Example

Example application:

    import sqlite3

    connection = sqlite3.connect("users.db")

    connection.execute("PRAGMA foreign_keys = ON")

    cursor = connection.cursor()

    # CREATE
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS users (
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            email TEXT UNIQUE,
            age INTEGER
        )
    """)

    # INSERT
    cursor.execute("""
        INSERT INTO users (name, email, age)
        VALUES (?, ?, ?)
    """, ("Ali", "ali@example.com", 25))

    connection.commit()

    # READ
    cursor.execute("SELECT * FROM users")

    users = cursor.fetchall()

    for user in users:
        print(user)

    # UPDATE
    cursor.execute("""
        UPDATE users
        SET age = ?
        WHERE id = ?
    """, (26, 1))

    connection.commit()

    # DELETE
    cursor.execute("""
        DELETE FROM users
        WHERE id = ?
    """, (1,))

    connection.commit()

    connection.close()

# 58. Better Python SQLite Pattern

For real applications, separate database operations into functions.

Example:

    import sqlite3

    DB_NAME = "app.db"


    def get_connection():
        connection = sqlite3.connect(DB_NAME)
        connection.row_factory = sqlite3.Row
        connection.execute("PRAGMA foreign_keys = ON")
        return connection


    def create_user(name, email):
        connection = get_connection()

        try:
            connection.execute(
                """
                INSERT INTO users (name, email)
                VALUES (?, ?)
                """,
                (name, email)
            )

            connection.commit()

        finally:
            connection.close()


    def get_users():
        connection = get_connection()

        try:
            cursor = connection.execute(
                "SELECT * FROM users"
            )

            return cursor.fetchall()

        finally:
            connection.close()

# 67. SQLite Security

Important practices:

### 1. Use parameterized queries

    cursor.execute(
        "SELECT * FROM users WHERE email = ?",
        (email,)
    )

### 2. Do not expose the database file publicly

Don't put:

    app.db

inside a publicly accessible static directory.

### 3. Protect sensitive data

SQLite does not automatically encrypt database files.

If sensitive information must be stored, consider appropriate encryption/security architecture.

### 4. Validate input

Database constraints are useful, but application-level validation is also important.

### 5. Back up the database

A database file should not be treated as disposable if it contains important data.

---

# 68. SQLite Limitations

SQLite is powerful, but it is not ideal for every application.

### Major limitations/considerations

- Limited concurrent writing compared with server databases
- No built-in client/server architecture
- Not ideal for very large high-concurrency applications
- Database-level security/access control is more limited than server databases
- Some advanced database features differ from PostgreSQL/MySQL
- Application architecture matters significantly for concurrent access

# 71. Important SQLite File Concepts

A SQLite database may involve files depending on configuration and journaling mode.

For example:

    app.db

During certain operations, SQLite may create auxiliary files related to journaling or WAL.

Therefore, don't assume the database always consists of only one physical file during active operation.

For backups, use SQLite-aware backup approaches when appropriate rather than blindly copying files during unsafe moments.

# 72. Common SQLite Errors

## "database is locked"

Usually means another operation currently holds a conflicting lock.

Possible solutions:

- Keep transactions short
- Close unused connections
- Configure timeout
- Consider WAL for suitable workloads
- Avoid unnecessary concurrent writes

---

## "UNIQUE constraint failed"

Example:

    email TEXT UNIQUE

Trying to insert an existing email can produce this error.

---

## "FOREIGN KEY constraint failed"

The referenced parent record does not satisfy the foreign-key relationship.

---

## "no such table"

Usually means:

- Table wasn't created
- Wrong database file
- Wrong working directory
- Connection points to another database

1. Use parameterized queries.
2. Use transactions for related operations.
3. Keep transactions short.
4. Enable foreign keys when relationships require enforcement.
5. Add indexes based on actual query patterns.
6. Don't create unnecessary indexes.
7. Validate application input.
8. Use constraints.
9. Close connections properly.
10. Use context managers where appropriate.
11. Keep database logic separate from UI/business logic.
12. Back up important databases.
13. Don't commit database files containing sensitive data to Git.
14. Add `*.db` to `.gitignore` when appropriate.
15. Use PostgreSQL or another server database when workload requirements exceed SQLite's strengths.

Usually don't commit local SQLite databases to Git if they contain:

- Temporary data
- Large generated datasets
- Environment-specific state

# 78. SQLite vs JSON

JSON is useful for hierarchical/document-style data.

SQLite is better when you need:

- Relationships
- SQL queries
- Constraints
- Transactions
- Indexing
- Structured relational data

JSON is often useful for:

- Configuration
- API payloads
- Nested documents
- Flexible structures

# 82. Quick Revision

### SQLite

> Lightweight, serverless, embedded relational database.

### Database

Usually stored in a file:

    app.db

### Python

    import sqlite3

### Connect

    connection = sqlite3.connect("app.db")

### Cursor

    cursor = connection.cursor()

### Execute

    cursor.execute(...)

### Save

    connection.commit()

### Undo uncommitted transaction

    connection.rollback()

### Read

    cursor.fetchone()
    cursor.fetchall()

### Foreign keys

    PRAGMA foreign_keys = ON

### In-memory database

    sqlite3.connect(":memory:")

### WAL

    PRAGMA journal_mode = WAL;

### Parameterized query

    cursor.execute(
        "SELECT * FROM users WHERE id = ?",
        (user_id,)
    )
