# 🛠️ Tools in LangGraph — Complete Detailed Notes

## 1. Introduction

Tools are one of the most important concepts in LangGraph because they allow an LLM-based application to interact with the outside world.

An LLM is very good at:

- Understanding natural language
- Reasoning over information
- Generating text
- Deciding what action may be useful
- Selecting an appropriate tool

However, an LLM by itself does not automatically have the ability to:

- Query your database
- Call your APIs
- Search live information
- Read application data
- Execute business operations
- Send emails
- Perform reliable calculations
- Retrieve information from external systems

Tools solve this problem.

A tool is essentially a function that an AI system can call to perform a specific operation.

The basic idea is:

    User
      │
      ▼
    LLM
      │
      ├── Does it need an external action?
      │
      ├── No ───────────────► Final Answer
      │
      └── Yes
           │
           ▼
        Select Tool
           │
           ▼
        Execute Tool
           │
           ▼
       Tool Result
           │
           ▼
          LLM
           │
           ▼
       Final Answer

The most important principle is:

> **The LLM decides what to do; the tool performs the actual operation.**

---

# 2. What Is a Tool?

A **Tool** is a callable function exposed to an LLM or AI Agent so that the agent can perform a specific operation.

For example, a normal Python function is:

    def add(a, b):
        return a + b

We can convert it into a LangChain tool:

    from langchain_core.tools import tool

    @tool
    def add(a: int, b: int) -> int:
        """Add two numbers."""
        return a + b

Now the tool contains information that the LLM can understand.

    Tool Name:
        add

    Description:
        Add two numbers.

    Input:
        a → integer
        b → integer

    Output:
        integer

The LLM can reason:

    "The user wants me to calculate 25 + 50.
     I should use the add tool."

The framework can then execute:

    add(25, 50)

Result:

    75

---

# 3. Tools vs Normal Functions

A normal Python function is designed primarily for programmers.

Example:

    def calculate_tax(amount, rate):
        return amount * rate

A tool is designed so that an LLM can understand and invoke the function.

Example:

    from langchain_core.tools import tool

    @tool
    def calculate_tax(amount: float, rate: float) -> float:
        """Calculate tax based on amount and tax rate."""
        return amount * rate

The important difference is not that the underlying logic must be completely different.

The important difference is that a tool provides metadata that can be exposed to the model.

Conceptually:

    Normal Function
        │
        ├── Function name
        ├── Parameters
        └── Implementation

    Tool
        │
        ├── Name
        ├── Description
        ├── Input schema
        ├── Function implementation
        └── Output

---

# 4. Anatomy of a Tool

A tool can be understood using five major components.

    Tool
     │
     ├── Name
     │
     ├── Description
     │
     ├── Input Schema
     │
     ├── Function Logic
     │
     └── Output

## 4.1 Tool Name

The tool name identifies the operation.

Example:

    search_database

    calculate_sum

    get_weather

    send_email

    retrieve_documents

The name should clearly describe what the tool does.

Bad:

    tool1

Good:

    search_customer

---

# 5. Tool Description

The description tells the LLM what the tool does and helps it decide when to use it.

Example:

    @tool
    def search_customer(customer_id: int) -> str:
        """Search for a customer using their customer ID."""
        return f"Customer {customer_id} found."

The description:

    Search for a customer using their customer ID.

is important because the model uses tool descriptions when deciding which tool is appropriate.

A vague description can cause poor tool selection.

Bad description:

    "Does something."

Better description:

    "Search for a customer using their unique customer ID."

---

# 6. Input Schema

A tool can define the expected inputs.

Example:

    @tool
    def multiply(a: int, b: int) -> int:
        """Multiply two integers."""
        return a * b

The expected schema is conceptually:

    multiply
      │
      ├── a: integer
      └── b: integer

The model can generate a tool call such as:

    {
        "name": "multiply",
        "arguments": {
            "a": 10,
            "b": 5
        }
    }

The framework then executes the underlying function.

Result:

    50

---

# 7. Tool Output

A tool produces a result.

Example:

    @tool
    def get_user_name(user_id: int) -> str:
        """Get the user's name using their ID."""
        return "Musharraf"

Input:

    user_id = 101

Output:

    Musharraf

The output is then returned to the agent.

The agent can use that information for further reasoning.

---

# 8. Basic Tool Creation

The simplest approach is using the `@tool` decorator.

Example:

    from langchain_core.tools import tool

    @tool
    def multiply(a: int, b: int) -> int:
        """Multiply two numbers."""
        return a * b

This creates a tool named:

    multiply

The function accepts:

    a
    b

and returns:

    a * b

# 9. Calling a Tool Directly

A tool can also be called directly from Python.

Example:

    result = multiply.invoke({
        "a": 10,
        "b": 5
    })

    print(result)

Output:

    50

The important point is that `.invoke()` is a standard way of executing a tool.

Conceptually:

    Tool
      │
      ▼
    invoke()
      │
      ▼
    Function
      │
      ▼
    Result

# 10. Tool Calling by an LLM

The more interesting use case is when the LLM decides to call the tool.

Suppose we have:

    @tool
    def multiply(a: int, b: int) -> int:
        """Multiply two numbers."""
        return a * b

We provide the tool to an LLM.

Conceptually:

    User:
    "What is 25 multiplied by 4?"

    ↓

    LLM:
    "I should use the multiply tool."

    ↓

    Tool Call:
    multiply(a=25, b=4)

    ↓

    Tool Execution:
    100

    ↓

    LLM:
    "25 multiplied by 4 is 100."

---

# 11. Tool Calling Is Not the Same as Tool Execution

This is a very important interview concept.

## Tool Calling

The LLM decides:

    "I want to call the multiply tool with:
     a = 25
     b = 4"

This is a **tool call**.

## Tool Execution

The application actually executes:

    multiply(25, 4)

This is **tool execution**.

Therefore:

    LLM
      │
      │ Tool Call
      ▼
    Application
      │
      │ Tool Execution
      ▼
    Tool
      │
      ▼
    Result

The LLM does not magically execute your Python function.

The application/framework is responsible for executing the tool.

---

# 12. What Is Tool Calling?

**Tool calling** is the capability where an LLM generates a structured request asking the application to execute a particular tool.

For example:

    User:
    "Calculate 25 × 10."

The model may generate conceptually:

    {
        "tool": "multiply",
        "arguments": {
            "a": 25,
            "b": 10
        }
    }

The application sees the request and executes:

    multiply(25, 10)

Result:

    250

The result is then supplied back to the model.

---

# 15. The ToolNode

LangGraph provides `ToolNode` for executing tools inside a graph.

It is commonly imported as:

    from langgraph.prebuilt import ToolNode

The purpose of `ToolNode` is to execute tool calls produced by the model.

Conceptually:

    LLM Node
       │
       │ AIMessage containing tool call
       ▼
    ToolNode
       │
       ├── Tool 1
       ├── Tool 2
       ├── Tool 3
       └── Tool 4
       │
       ▼
    ToolMessage
       │
       ▼
    LLM Node

This greatly simplifies building tool-using agents.

---

# 16. Basic ToolNode Example

First create a tool:

    from langchain_core.tools import tool

    @tool
    def add(a: int, b: int) -> int:
        """Add two numbers."""
        return a + b

Create the ToolNode:

    from langgraph.prebuilt import ToolNode

    tools = [add]

    tool_node = ToolNode(tools)

Now the ToolNode knows about the available tools.

---

# 17. LLM + Tools

The LLM must know which tools are available.

Conceptually:

    tools = [
        add,
        multiply,
        search_database
    ]

Then tools are bound to the model.

For supported chat models, the pattern is generally:

    model_with_tools = model.bind_tools(tools)

The model can now produce tool calls when appropriate.

Example:

    from langchain_openai import ChatOpenAI

    model = ChatOpenAI(
        model="gpt-4.1-mini"
    )

    model_with_tools = model.bind_tools(tools)

The exact model name and API configuration can change over time, but the architectural idea remains the same.

---

# 18. Important: bind_tools()

`bind_tools()` connects tool definitions to the model.

Conceptually:

    Model
      +
    Tools
      ↓
    Model with Tool-Calling Capability

Example:

    model_with_tools = model.bind_tools([
        add,
        multiply
    ])

This does not mean that the model itself executes the Python functions.

It means that the model is given information about the available tools and can produce structured tool calls.

The application then executes those calls.

# 24. Why Tool Call IDs Matter

A model may make multiple tool calls.

For example:

    Call 1:
        get_weather("Mumbai")

    Call 2:
        get_weather("Delhi")

The system needs to know which result belongs to which tool call.

Tool call IDs help maintain this relationship.

Conceptually:

    AIMessage
       │
       ├── call_001 → Mumbai
       └── call_002 → Delhi

    ToolMessage
       │
       ├── call_001 → Mumbai weather result
       └── call_002 → Delhi weather result

This becomes especially important when tools are executed in parallel.

# 28. Tool Routing

In LangGraph, routing determines where the graph should go next.

Example:

    LLM
     │
     ▼
    Does AIMessage contain tool calls?
       /              \
     Yes               No
      │                 │
      ▼                 ▼
    ToolNode            END

This can be implemented using conditional edges.

Conceptually:

    def should_continue(state):
        last_message = state["messages"][-1]

        if last_message.tool_calls:
            return "tools"

        return "end"

Then:

    graph.add_conditional_edges(
        "llm",
        should_continue,
        {
            "tools": "tools",
            "end": END
        }
    )

This creates the agent loop.

# 30. Complete Simple Example

The following is a simplified example of a LangGraph tool-using workflow.

    from typing import Annotated
    from typing_extensions import TypedDict

    from langchain_core.messages import HumanMessage
    from langchain_core.tools import tool
    from langchain_openai import ChatOpenAI

    from langgraph.graph import StateGraph, START, END
    from langgraph.graph.message import add_messages
    from langgraph.prebuilt import ToolNode

    class State(TypedDict):
        messages: Annotated[list, add_messages]

    @tool
    def multiply(a: int, b: int) -> int:
        """Multiply two integers."""
        return a * b

    tools = [multiply]

    model = ChatOpenAI(
        model="gpt-4.1-mini"
    )

    model_with_tools = model.bind_tools(tools)

    def call_model(state: State):
        response = model_with_tools.invoke(state["messages"])
        return {
            "messages": [response]
        }

    tool_node = ToolNode(tools)

    def should_continue(state: State):
        last_message = state["messages"][-1]

        if getattr(last_message, "tool_calls", None):
            return "tools"

        return END

    graph = StateGraph(State)

    graph.add_node("llm", call_model)
    graph.add_node("tools", tool_node)

    graph.add_edge(START, "llm")

    graph.add_conditional_edges(
        "llm",
        should_continue,
        {
            "tools": "tools",
            END: END
        }
    )

    graph.add_edge("tools", "llm")

    app = graph.compile()

    result = app.invoke({
        "messages": [
            HumanMessage(
                content="What is 25 multiplied by 10?"
            )
        ]
    })

The important architecture is:

    START
      ↓
    LLM
      ↓
    Tool Call
      ↓
    ToolNode
      ↓
    LLM
      ↓
    Final Answer

# 33. Agent Loop

This pattern is commonly called the **agent loop**.

The loop is:

    Reason
      ↓
    Act
      ↓
    Observe
      ↓
    Reason
      ↓
    Act
      ↓
    Observe
      ↓
    ...
      ↓
    Final Answer

In terms of LangGraph:

    LLM Node
       ↓
    Tool Node
       ↓
    LLM Node
       ↓
    Tool Node
       ↓
    LLM Node
       ↓
      END

# 34. Tools and ReAct-Style Agents

A classic agent pattern is often described as:

    Reason
      ↓
    Act
      ↓
    Observe
      ↓
    Reason
      ↓
    Act
      ↓
    Observe

For example:

    User:
    "Find the weather in Mumbai and tell me whether I need an umbrella."

The agent may:

    Reason:
    I need current weather.

    Act:
    Call weather tool.

    Observe:
    Rain is expected.

    Reason:
    Rain is expected, so the user may need an umbrella.

    Final Answer:
    "Rain is expected, so you should carry an umbrella."

Modern systems may implement this loop through structured tool calling rather than exposing the model's private chain-of-thought.

# 35. Tool Calling vs Agent

These concepts are related but not identical.

## Tool Calling

The model calls a predefined function.

Example:

    User
      ↓
    LLM
      ↓
    calculator
      ↓
    Result

## Agent

An agent can repeatedly reason, choose tools, inspect results, and decide what to do next.

Example:

    User
      ↓
    Agent
      ↓
    Search Tool
      ↓
    Result
      ↓
    Agent
      ↓
    Calculator
      ↓
    Result
      ↓
    Agent
      ↓
    Final Answer

Therefore:

> **Tool calling is a capability; an agent is a system that can use capabilities to achieve a goal.**

# 37. Parallel Tool Execution

Suppose an agent needs:

    Weather(Mumbai)
    Weather(Delhi)
    Weather(Pune)

If the operations are independent, they can potentially be executed concurrently.

Conceptually:

    LLM
     │
     ├────────► Weather(Mumbai)
     │
     ├────────► Weather(Delhi)
     │
     └────────► Weather(Pune)
                   │
                   ▼
              Combined Results
                   │
                   ▼
                  LLM

Parallel execution can reduce latency when tools are independent.

However, whether calls actually execute in parallel depends on the runtime, tool implementation, and graph design.

# 48. Tool Error Handling

Tools can fail.

Examples:

- API unavailable
- Invalid arguments
- Database connection failure
- Authentication failure
- Timeout
- Missing file
- Rate limit
- Invalid user input

Example:

            raise ValueError("Cannot divide by zero.")

        return a / b

      ┌─────┘
      │

       │
       ▼

       │

# 50. Retry Strategy

Some tool failures are temporary.

For example:

    API Timeout
       ↓
    Retry
       ↓
    API Success

But retries should be controlled.

Do not blindly retry every operation.

For example, retrying a read operation may be safe.

Retrying an operation such as:

    charge_credit_card()

could potentially cause duplicate side effects if the first operation actually succeeded but the response was lost.

Therefore:

> **Retries must consider whether the tool is read-only or side-effecting.**

# 51. Read-Only vs Side-Effecting Tools

This is an important production concept.

## Read-Only Tool

A read-only tool retrieves information.

Examples:

    search_database()

    get_weather()

    get_customer()

    retrieve_documents()

These generally do not change external state.

## Side-Effecting Tool

A side-effecting tool changes something.

Examples:

    send_email()

    delete_customer()

    create_order()

    transfer_money()

    update_database()

These require greater caution.

# 53. Tool Security

Tools are powerful because they allow agents to interact with real systems.

That also creates security risks.

Potential risks include:

- Unauthorized actions
- Prompt injection
- Data leakage
- Excessive permissions
- Malicious tool arguments
- SQL injection
- Arbitrary file access
- Dangerous API calls
- Credential exposure

Therefore, tools should follow the principle of:

> **Least privilege**

Give a tool only the permissions it actually needs.

---

# 54. Example of Least Privilege

Suppose an agent only needs to read customer information.

Bad design:

    database_tool
        permissions:
            SELECT
            INSERT
            UPDATE
            DELETE
            DROP

Better design:

    customer_read_tool
        permissions:
            SELECT only

The agent should not receive unnecessary capabilities.

# 55. Validate Tool Inputs

Never blindly trust tool arguments.

Suppose:

        amount: float
    ):
        ...

The system should validate:

- Account format
- Amount range
- User authorization
- Currency
- Transaction limits
- Confirmation requirements

Tools should have defensive validation.

Better:

# 59. Tool Granularity

Too many tools:

# 80. Tool Design Principles

Good tools generally follow these principles:

### 1. Single Responsibility

One tool should have one clear purpose.

### 2. Clear Description

The model should easily understand when to use it.

### 3. Strong Input Validation

Never trust arbitrary model-generated arguments.

### 4. Structured Input

Use typed parameters where possible.

### 5. Controlled Output

Return concise, useful results.

### 6. Least Privilege

Give the tool only necessary permissions.

### 7. Error Handling

Handle failures gracefully.

### 8. Observability

Log and trace tool calls.

### 9. Idempotency Where Possible

Repeated execution should not accidentally duplicate side effects.

### 10. Human Approval for High-Impact Actions

Require confirmation when appropriate.

# 81. Idempotency

Idempotency is important for side-effecting tools.

An operation is idempotent when repeating it produces the same intended final state rather than creating duplicate effects.

For example:

    set_user_status("active")

may be idempotent.

But:

    charge_card(1000)

may not be idempotent.

If the agent retries:

    charge_card(1000)
    charge_card(1000)

the customer could potentially be charged twice.

Therefore, financial and other side-effecting tools need special safeguards.

This is the foundation of tool-using agents.

# 95. Interview Question: What Is a Tool?

### Answer

A tool is a callable function that an LLM or agent can invoke to perform an operation that goes beyond the model's native capabilities.

For example, a weather tool can call a weather API and return current weather information.

In LangGraph, tools can be integrated into graph workflows using components such as `ToolNode`, allowing the model to decide when a tool should be called and enabling the graph to execute the tool and return its result back to the model.

---

# 96. Interview Question: What Is Tool Calling?

### Answer

Tool calling is the capability of an LLM to generate a structured request specifying which tool should be executed and what arguments should be passed.

The LLM does not necessarily execute the tool itself. Instead, the application or framework receives the tool call, executes the corresponding function, and returns the result to the model.

---

# 97. Interview Question: What Is ToolNode?

### Answer

`ToolNode` is a prebuilt LangGraph node designed to execute tools specified by tool calls in the model's messages.

It simplifies the implementation of tool execution within a LangGraph workflow.

A common architecture is:

    LLM
      ↓
    ToolNode
      ↓
    Tool Result
      ↓
    LLM

---

# 98. Interview Question: What Is bind_tools()?

### Answer

`bind_tools()` associates tool definitions with a chat model that supports tool calling.

It allows the model to know about the available tools and generate structured tool calls when it determines that a tool is required.

However, binding tools does not mean the model directly executes those Python functions. The application or runtime executes the generated tool calls.

---

# 99. Interview Question: What Is the Difference Between a Tool and an Agent?

### Answer

A tool is a specific capability or function.

An agent is a system that can reason about a task, decide which actions are required, invoke tools, inspect their results, and continue until the task is completed.

For example:

    Calculator = Tool

    AI system that decides:
    "I need the calculator, then I need the search tool, then I can answer."

    = Agent

---

# 100. Interview Question: Why Are Tools Important for Agentic AI?

### Answer

Tools allow agents to interact with external systems and take meaningful actions.

Without tools, an LLM is primarily limited to generating responses.

With tools, an agent can:

- Retrieve live information
- Query databases
- Call APIs
- Search documents
- Perform calculations
- Execute business operations
- Interact with applications

Therefore, tools are a fundamental mechanism for connecting LLM reasoning with real-world actions.

---

# 101. Interview Question: How Does a LangGraph Agent Use a Tool?

### Answer

A typical LangGraph agent follows this process:

    1. User sends a request.
    2. LLM analyzes the request.
    3. LLM decides whether a tool is required.
    4. If required, the LLM generates a tool call.
    5. LangGraph routes the call to a ToolNode.
    6. ToolNode executes the tool.
    7. Tool result is added to the message/state.
    8. The result is sent back to the LLM.
    9. The LLM decides whether another tool is required.
    10. The process repeats until the final answer is produced.

# 105. Final Summary

Tools are a fundamental building block of LangGraph-based Agentic AI systems.

The key concepts are:

- A tool is a callable function.
- Tools extend an LLM's capabilities.
- Tools can interact with APIs, databases, files, search systems, and applications.
- `@tool` can be used to create tools.
- Tool descriptions help the LLM select the correct tool.
- Input schemas define what arguments a tool accepts.
- `bind_tools()` connects tools to a compatible chat model.
- The LLM generates a structured tool call.
- The application/runtime executes the tool.
- `ToolNode` can execute tool calls inside LangGraph.
- Tool results are returned as messages.
- `AIMessage` can contain tool calls.
- `ToolMessage` can contain tool results.
- Conditional edges can route the graph based on tool calls.
- Tool execution can create an agent loop.
- Multiple tools allow agents to solve more complex tasks.
- Independent tools may be executed in parallel.
- Dependent tools may need sequential execution.
- Tools can be used for RAG, databases, APIs, calculations, files, and business systems.
- Tool errors must be handled.
- Tool inputs must be validated.
- Tools should follow least-privilege security.
- Side-effecting tools require additional safeguards.
- Human approval can be added for sensitive operations.
- Tool execution should be observable in production.
- Good tool design is critical for reliable Agentic AI.
