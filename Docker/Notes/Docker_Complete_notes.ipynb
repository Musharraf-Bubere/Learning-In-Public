# 1. What is Docker?

Docker is an open-source platform used to **build, package, distribute, and run applications inside isolated environments called containers**.

The main purpose of Docker is to make an application environment reproducible and portable.

Consider a Python application that requires:

    Python 3.11
    FastAPI
    Pandas
    NumPy
    Scikit-learn
    Specific package versions

On the developer's machine everything works correctly.

However, when the application is moved to another machine, the environment might be different:

    Python 3.9
    Different package versions
    Different operating system
    Missing system libraries
    Different environment variables

The application can therefore behave differently.

This leads to the common statement:

    "It works on my machine!"

Docker solves much of this environment inconsistency by packaging the application and its required environment into a Docker image.

The basic workflow is:

    Application
         +
    Dependencies
         +
    Runtime
         +
    Configuration
         ↓
    Dockerfile
         ↓
    Docker Image
         ↓
    Docker Container
         ↓
    Running Application

A useful definition for interviews is:

    Docker is a containerization platform that packages an application
    and its dependencies into a portable image that can be executed
    consistently in isolated containers.

# 3. Problems Before Docker

Before containers became common, applications were commonly deployed directly on servers or inside virtual machines.

Imagine a server hosting multiple applications:

    Server
    │
    ├── Application A
    │   └── Python 3.9
    │
    ├── Application B
    │   └── Python 3.11
    │
    └── Application C
        └── Node.js

Now suppose:

    Application A
        requires package X version 1

    Application B
        requires package X version 2

This can create dependency conflicts.

Docker allows these applications to run in separate containers.

    Host Machine
    │
    ├── Container A
    │   └── Application A
    │       └── Dependencies
    │
    ├── Container B
    │   └── Application B
    │       └── Dependencies
    │
    └── Container C
        └── Application C
            └── Dependencies

# 4. Main Benefits of Docker

Docker provides several important benefits.

## 4.1 Portability

A Docker image can be moved between compatible environments.

For example:

    Developer Machine
          ↓
    Docker Image
          ↓
    Testing Server
          ↓
    Production Server

The application environment is packaged consistently.

---

## 4.2 Reproducibility

If an image is built from a defined Dockerfile and dependency configuration, another developer can recreate the same environment.

This is extremely useful in teams.

---

## 4.3 Isolation

Applications can run inside separate containers.

For example:

    Container 1
    └── FastAPI Application

    Container 2
    └── PostgreSQL

    Container 3
    └── Redis

They can communicate through Docker networking while remaining isolated processes.

---

## 4.4 Scalability

A single image can be used to create multiple containers.

    Image
      │
      ├── Container 1
      ├── Container 2
      ├── Container 3
      └── Container 4

This is useful for scaling applications.

---

## 4.5 Faster Deployment

Instead of manually installing the complete environment on a server, we can pull an image and run it.

    Registry
       ↓
    docker pull
       ↓
    Image
       ↓
    docker run
       ↓
    Application

---

## 4.6 Consistent Development Environment

A team can use the same containerized environment.

    Developer A
        ↓
    Docker

    Developer B
        ↓
    Docker

    Developer C
        ↓
    Docker

This reduces "works on my machine" problems.

# 6. Dockerfile

A Dockerfile is a text file containing instructions used to build a Docker image.

Example:

    FROM python:3.11-slim

    WORKDIR /app

    COPY requirements.txt .

    RUN pip install --no-cache-dir -r requirements.txt

    COPY . .

    CMD ["python", "app.py"]

The Dockerfile describes how the environment should be constructed.

Conceptually:

    Dockerfile
        │
        ├── Base image
        ├── Working directory
        ├── Dependencies
        ├── Application files
        └── Startup command
              ↓
          Docker Image

---

# 7. Docker Image

A Docker image is a packaged, read-only template used to create containers.

An image may contain:

    Application code
    Python runtime
    Python libraries
    System libraries
    Configuration defaults
    Startup instructions

For example:

    Python Base Image
           +
    FastAPI
           +
    Requirements
           +
    Application Code
           ↓
    fastapi-app:1.0

The image itself is not normally the running application.

It is the template from which containers are created.

---

# 8. Docker Container

A container is a running or stopped instance of a Docker image.

Example:

    Image:

        my-fastapi-app:1.0

    Run:

        docker run my-fastapi-app:1.0

    Result:

        Container

Relationship:

    One Image
       │
       ├── Container A
       ├── Container B
       └── Container C

Multiple containers can be created from the same image.

# 16. Containers and the Host Operating System

Containers are not full virtual machines.

They use operating-system-level isolation.

A simplified conceptual architecture is:

    Host Operating System
    │
    ├── Container Runtime
    │
    ├── Container A
    │   └── Application A
    │
    ├── Container B
    │   └── Application B
    │
    └── Container C
        └── Application C

On Linux, containers share the host kernel.

Docker Desktop on Windows and macOS uses a Linux environment/virtualization layer to run Linux containers because Linux containers require a Linux kernel environment.

---

# 17. Containers vs Virtual Machines

A virtual-machine architecture looks like:

    Physical Hardware
          ↓
       Host OS
          ↓
      Hypervisor
       /      \
      ↓        ↓
     VM 1     VM 2
      │        │
    Guest OS Guest OS
      │        │
    App A     App B

A container architecture looks like:

    Physical Hardware
          ↓
       Host OS
          ↓
    Container Runtime
       /       \
      ↓         ↓
 Container A  Container B
      │         │
    App A      App B

Containers generally require fewer resources because they do not each need a complete guest operating system.

---

# 18. Docker vs Virtual Machine

| Feature | Docker Container | Virtual Machine |
|---|---|---|
| Isolation | Process-level/OS-level | Full machine virtualization |
| Guest OS | Usually no separate full OS | Yes |
| Startup | Usually fast | Usually slower |
| Resource usage | Usually lower | Usually higher |
| Application density | High | Lower |
| Portability | High | High |
| Full OS isolation | No | Yes |
| Common use | Microservices, APIs, workers | Full OS environments |

Docker is not simply a replacement for virtual machines.

They can also be used together:

    Physical Server
         ↓
       VM
         ↓
      Docker
         ↓
    Containers

# 20. FROM

`FROM` defines the base image.

Example:

    FROM python:3.11-slim

This tells Docker:

    Start building this image from
    the Python 3.11 slim image.

Every Dockerfile normally begins with a `FROM` instruction unless using a special build pattern.

---

# 21. Base Image

A base image provides the starting point for your application image.

For example:

    python:3.11-slim

provides a Python environment.

Other examples:

    ubuntu
    debian
    alpine
    node
    nginx

For Python applications, a common starting point is:

    python:3.11-slim

---

# 22. Why Use Slim Images?

A slim image contains fewer unnecessary components than a full distribution image.

For example:

    python:3.11

may be larger than:

    python:3.11-slim

Smaller images can provide:

    Faster downloads
    Faster deployments
    Smaller storage requirements
    Smaller attack surface

However, slim images may require additional system packages for some Python dependencies.

Therefore, image selection is a tradeoff.

---

# 23. Alpine Images

Another common option is Alpine Linux.

Example:

    python:3.11-alpine

Alpine images are often small.

However, some Python packages can be more difficult to install on Alpine because Alpine uses musl libc rather than glibc and may require compilation tools.

Therefore:

    Smallest image
        ≠
    Always best image

For many Python applications, `python:3.11-slim` is a practical starting point.

---

# 24. WORKDIR

`WORKDIR` defines the working directory inside the image.

Example:

    WORKDIR /app

After this instruction, subsequent commands operate relative to `/app`.

Example:

    WORKDIR /app

    COPY . .

The application files are copied into:

    /app

Using `WORKDIR` is preferable to repeatedly using `cd`.

---

# 25. COPY

`COPY` copies files from the Docker build context into the image.

Example:

    COPY requirements.txt .

This copies:

    Host:
        requirements.txt

to:

    Container image:
        /app/requirements.txt

if the current `WORKDIR` is `/app`.

Another example:

    COPY . .

copies the build-context files into the working directory.

---

# 26. ADD

`ADD` can copy files and supports additional behaviors such as extracting local archives.

Example:

    ADD application.tar.gz /app/

For normal file copying, `COPY` is generally preferred because it is simpler and more explicit.

---

# 27. RUN

`RUN` executes a command while building the image.

Example:

    RUN pip install --no-cache-dir -r requirements.txt

This happens during:

    docker build

not when the container starts.

Remember:

    RUN
    → Build time

---

# 28. CMD

`CMD` defines the default command executed when a container starts.

Example:

    CMD ["python", "app.py"]

If the container starts normally:

    docker run myapp

Docker runs:

    python app.py

Important:

    CMD
    → Runtime default command

---

# 29. ENTRYPOINT

`ENTRYPOINT` defines the primary executable of the container.

Example:

    ENTRYPOINT ["python"]

Combined with:

    CMD ["app.py"]

the effective command becomes:

    python app.py

ENTRYPOINT is useful when the container represents a specific executable.

---

# 30. CMD vs ENTRYPOINT

| CMD | ENTRYPOINT |
|---|---|
| Default command/arguments | Main executable |
| Easier to override | More fixed |
| Provides defaults | Defines container purpose |
| Common for application startup | Useful for executable-style images |

Example:

    ENTRYPOINT ["python"]
    CMD ["app.py"]

Default:

    python app.py

If we run:

    docker run myapp other.py

the command becomes conceptually:

    python other.py

---

# 31. EXPOSE

Example:

    EXPOSE 8000

This documents that the application listens on port 8000.

It does not automatically publish the port to the host.

To publish the port:

    docker run -p 8000:8000 myapp

The distinction is:

    EXPOSE
    → Documentation/metadata

    -p
    → Host-to-container port publishing

---

# 32. ENV

`ENV` defines environment variables.

Example:

    ENV APP_ENV=production

Inside the container:

    APP_ENV
        ↓
    production

Environment variables are useful for non-secret configuration.

Do not store sensitive credentials directly in a Dockerfile.

Bad practice:

    ENV API_KEY=secret-value

Use runtime secrets/environment configuration instead.

---

# 33. ARG

`ARG` defines a build-time variable.

Example:

    ARG PYTHON_VERSION=3.11

    FROM python:${PYTHON_VERSION}-slim

The important distinction is:

    ARG
    → Build time

    ENV
    → Runtime environment

Build arguments should not be treated as a secure mechanism for secrets.

---

# 34. USER

`USER` specifies the user used for subsequent commands and/or the container process.

Example:

    RUN useradd -m appuser

    USER appuser

Running applications as a non-root user is generally a better security practice when root privileges are unnecessary.

---

# 35. LABEL

`LABEL` adds metadata to an image.

Example:

    LABEL maintainer="team"
    LABEL version="1.0"

Labels can be used to organize and identify images.

---

# 36. HEALTHCHECK

`HEALTHCHECK` defines a command Docker can use to determine whether a container is healthy.

Example:

    HEALTHCHECK CMD curl --fail http://localhost:8000/health || exit 1

Conceptually:

    Container
       ↓
    Healthcheck
       ↓
    Healthy / Unhealthy

Healthchecks can be useful for services such as APIs.

# 37. Complete Dockerfile Example

A practical FastAPI Dockerfile:

    FROM python:3.11-slim

    WORKDIR /app

    ENV PYTHONDONTWRITEBYTECODE=1
    ENV PYTHONUNBUFFERED=1

    COPY requirements.txt .

    RUN pip install --no-cache-dir -r requirements.txt

    COPY . .

    EXPOSE 8000

    CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]

Explanation:

    FROM
    → Python runtime

    WORKDIR
    → Application directory

    ENV
    → Python runtime behavior

    COPY requirements.txt
    → Dependency file

    RUN
    → Install dependencies

    COPY . .
    → Application code

    EXPOSE
    → Documents port

    CMD
    → Starts FastAPI

# 38. Docker Build Context

When we run:

    docker build .

the final `.` represents the build context.

The Docker client sends files from the build context to the builder.

For example:

    project/
    │
    ├── Dockerfile
    ├── requirements.txt
    ├── app.py
    └── data/

If we run:

    docker build .

the project directory becomes the build context.

This is why `.dockerignore` is important.

---

# 39. .dockerignore

`.dockerignore` tells Docker which files should not be included in the build context.

Example:

    venv/
    __pycache__/
    *.pyc
    .git/
    .env
    .pytest_cache/
    *.log

For a Python project:

    venv/
    __pycache__/
    *.pyc
    .pytest_cache/
    .git/
    .env
    .env.*
    *.log

This improves:

    Build speed
    Build context size
    Security
    Image hygiene

---

# 40. Why .dockerignore Is Important

Suppose the project contains:

    project/
    │
    ├── venv/
    │   └── thousands of files
    │
    ├── .git/
    │   └── repository history
    │
    ├── .env
    │   └── secrets
    │
    └── app.py

If all of this is sent as build context, the build can become unnecessarily large and sensitive files could be exposed to build steps.

Therefore:

    Project
       ↓
    .dockerignore
       ↓
    Clean Build Context
       ↓
    Docker Build

# 41. Docker Image Layers

Docker images are constructed from layers.

Consider:

    FROM python:3.11-slim
              ↓
         Layer 1

    COPY requirements.txt .
              ↓
         Layer 2

    RUN pip install ...
              ↓
         Layer 3

    COPY . .
              ↓
         Layer 4

Conceptually:

    ┌──────────────────────┐
    │ Application Code     │
    ├──────────────────────┤
    │ Dependencies         │
    ├──────────────────────┤
    │ requirements.txt     │
    ├──────────────────────┤
    │ Python Base Image    │
    └──────────────────────┘

Layers are important because Docker can reuse unchanged build results.

---

# 42. Docker Build Cache

Suppose the Dockerfile is:

    FROM python:3.11-slim

    WORKDIR /app

    COPY requirements.txt .

    RUN pip install -r requirements.txt

    COPY . .

If only `app.py` changes, Docker can often reuse the earlier layers and rebuild the later application layer.

This makes builds faster.

---

# 43. Good Dockerfile Layer Ordering

A common best practice is:

    COPY dependency file
        ↓
    Install dependencies
        ↓
    COPY application code

Example:

    COPY requirements.txt .

    RUN pip install --no-cache-dir -r requirements.txt

    COPY . .

Why?

Because application code changes frequently, while dependencies may change less frequently.

This allows dependency layers to remain cached more often.

---

# 44. Bad Layer Ordering

Avoid unnecessarily doing:

    COPY . .

    RUN pip install -r requirements.txt

If one application source file changes, Docker may need to invalidate the dependency installation layer.

Better:

    COPY requirements.txt .

    RUN pip install -r requirements.txt

    COPY . .

# 51. Why 0.0.0.0 Is Important

Suppose FastAPI starts with:

    uvicorn main:app --host 127.0.0.1 --port 8000

Inside a container, binding only to loopback can prevent access from outside the container.

For containerized web applications, we commonly use:

    --host 0.0.0.0

Example:

    uvicorn main:app --host 0.0.0.0 --port 8000

This allows the application to listen on the container's network interfaces.

# 69. Why Configuration Should Be External

Suppose we build:

    myapp:1.0

# 70. Docker Volumes

Containers are often treated as ephemeral.

If data is stored only inside a container's writable layer, deleting the container can cause the data to disappear.

For persistent data, Docker provides volumes.

Architecture:

    Container
       ↓
    Volume
       ↓
    Persistent Data

Example:

    PostgreSQL Container
          ↓
    PostgreSQL Volume
          ↓
    Database Files

# 71. Why Volumes Are Needed

Consider PostgreSQL:

    PostgreSQL Container
         ↓
    Database Data

If the container is removed:

    Container deleted
         ↓
    Data may be lost

With a volume:

    PostgreSQL Container
         ↓
    Docker Volume
         ↓
    Database Data

Now the container can be recreated while the data remains in the volume.

# 106. Multi-Stage Builds

    Builder Stage
         ↓
    Compile/Install
         ↓
    Required Artifacts
         ↓
    Runtime Stage
         ↓

      ↓
    Compile
      ↓
    Copy required output
      ↓
    Runtime Image

The final image can therefore be cleaner.

# 138. localhost Inside a Container

        ↓

# 198. Interview Question: What is Docker?

Answer:

    Docker is a containerization platform that allows us to package
    applications and their dependencies into portable images and run
    them as isolated containers.

---

# 199. Interview Question: What is a Container?

Answer:

    A container is a running instance of a Docker image. It provides
    an isolated environment for executing an application while sharing
    the underlying host kernel in the typical container model.

---

# 200. Interview Question: What is a Docker Image?

Answer:

    A Docker image is a packaged, immutable template containing the
    application and the dependencies required to create containers.

---

# 201. Interview Question: What is a Dockerfile?

Answer:

    A Dockerfile is a text file containing instructions that define
    how a Docker image should be built.

---

# 202. Interview Question: Image vs Container?

Answer:

    An image is a template used to create containers, while a container
    is a runtime instance of an image.

    Image
      ↓
    Container

---

# 203. Interview Question: What is Docker Compose?

Answer:

    Docker Compose is a tool for defining and running multi-container
    applications using a YAML configuration file.

It is commonly used for local development and multi-service application environments.

---

# 204. Interview Question: What is Docker Hub?

Answer:

    Docker Hub is a container registry that allows users and
    organizations to store, distribute, and retrieve container images.

---

# 205. Interview Question: Why Use .dockerignore?

Answer:

    .dockerignore excludes unnecessary or sensitive files from the
    Docker build context. This can reduce build size, improve build
    performance, and prevent unwanted files from being included in
    the build context.

---

# 206. Interview Question: RUN vs CMD?

Answer:

    RUN executes commands while building the image.

    CMD specifies the default command that runs when the container starts.

---

# 207. Interview Question: CMD vs ENTRYPOINT?

Answer:

    CMD provides default runtime command or arguments, while ENTRYPOINT
    defines the primary executable of the container. They can be used
    together when the executable should remain fixed while default
    arguments remain configurable.

---

# 208. Interview Question: Why Use Docker Volumes?

Answer:

    Containers can be temporary, so important data should not depend
    only on the container's writable layer. Docker volumes provide
    persistent storage that can survive container recreation.

---

# 209. Interview Question: What is Docker Networking?

Answer:

    Docker networking provides communication between containers and
    external systems. Containers attached to the same user-defined
    network can communicate using service or container names.

---

# 210. Interview Question: Why use 0.0.0.0 in a Dockerized API?

Answer:

    Binding an application to 0.0.0.0 makes it listen on the container's
    network interfaces, allowing published ports to forward traffic
    to the application. Binding only to 127.0.0.1 can restrict access
    to the container's loopback interface.

---

# 211. Interview Question: What is a Multi-Stage Build?

Answer:

    A multi-stage Docker build uses multiple build stages so that
    build tools and intermediate artifacts can remain in an earlier
    stage while only the required runtime artifacts are copied into
    the final image.

This can produce smaller and cleaner production images.

# 212. Interview Question: Docker vs VM?

Answer:

    A virtual machine generally virtualizes a complete operating-system
    environment, while containers provide OS-level/process isolation
    and typically share the host kernel.

    Containers generally start faster and consume fewer resources,
    while VMs provide stronger full-machine isolation.

---

# 213. Interview Question: Does Docker Guarantee Security?

Answer:

    No.

    Docker provides isolation and security mechanisms, but a secure
    application requires additional practices such as least privilege,
    secure images, vulnerability scanning, secret management, network
    controls, updates, and proper host security.

# 214. Interview Question: How Would You Debug a Docker Container?

Answer:

    I would first check whether the container is running using:

        docker ps

    Then I would inspect all containers:

        docker ps -a

    Next I would inspect logs:

        docker logs <container>

    If necessary, I would enter the container:

        docker exec -it <container> sh

    I would also inspect configuration and resources using:

        docker inspect <container>
        docker stats

# 217. Interview Question: Why Should Secrets Not Be Stored in a Dockerfile?

Answer:

    Docker images are distributed artifacts and image layers can
    retain information from build steps. Therefore, hardcoding secrets
    in a Dockerfile can expose credentials.

    Secrets should instead be supplied through appropriate runtime
    secret-management mechanisms.

    system libraries, ML frameworks, and sometimes GPU-compatible
    environments.

    Docker packages these requirements into a reproducible environment,
    making model-serving and deployment more consistent across
    development, testing, and production environments.

# 223. Golden Rules of Docker

Remember these rules:

    1. Dockerfile builds the image.

    2. Image creates the container.

    3. Container runs the application.

    4. Registry stores and distributes images.

    5. Docker Compose manages multi-container applications.

    6. Volumes persist important data.

    7. Networks allow containers to communicate.

    8. EXPOSE does not publish a port.

    9. -p publishes a port.

    10. RUN executes during image build.

    11. CMD defines a default runtime command.

    12. ENTRYPOINT defines the main executable.

    13. ARG is primarily for build-time configuration.

    14. ENV provides environment configuration.

    15. Do not hardcode secrets into images.

    16. Use .dockerignore.

    17. Prefer non-root containers when practical.

    18. Optimize image layers and build cache.

    19. Use healthchecks where useful.

    20. Treat containers as replaceable and persistent data separately.
