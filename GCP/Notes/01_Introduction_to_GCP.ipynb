# GCP Step 1: Introduction to Cloud Computing & GCP

## What is Cloud Computing?
Renting computing power (CPU, storage, memory, networking) from a provider
(like Google) instead of buying/maintaining physical servers yourself.
Pay only for what you use; scale up/down in seconds.

## Service Models
| Model | You manage        | Google manages          | Example         |
|-------|--------------------|--------------------------|-----------------|
| IaaS  | OS, apps, data     | Physical servers, network| Compute Engine  |
| PaaS  | Just app code      | Servers, OS, runtime     | App Engine, Cloud Run |
| SaaS  | Nothing            | Everything               | Gmail, Docs     |

Analogy: IaaS = rent kitchen & cook yourself. PaaS = get dough + toppings, bake it.
SaaS = pizza delivered ready to eat.

## GCP Global Infrastructure Hierarchy
Global Network
  -> Region (geographic area, e.g. asia-south1 = Mumbai)
       -> Zone (isolated data center within a region, e.g. asia-south1-a)

- Zones are isolated from each other (failure in one doesn't affect others)
- Pick a region close to your users for lower latency
- Spread critical resources across multiple zones for high availability

## Practical: Account Setup
1. Sign up at console.cloud.google.com (Google account required)
2. Free trial: $300 credit, ~90 days, no auto-charge
3. Create first project: Project dropdown -> New Project -> name it
4. Open Cloud Shell (>_ icon, top right) - free terminal, gcloud CLI preinstalled
5. Verify setup by running these commands in Cloud Shell:
   gcloud config list
   gcloud projects list

## Key Terms to Remember
- Project: container for all your GCP resources (billing, permissions, APIs)
- Region: geographic location (e.g. asia-south1)
- Zone: isolated data center inside a region (e.g. asia-south1-a)
- gcloud: GCP's command-line tool
- Cloud Shell: free browser-based terminal with gcloud preinstalled
