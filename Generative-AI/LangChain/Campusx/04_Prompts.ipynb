# What are Prompts?

## Definition

**Prompts are the input instructions or queries given to a model to guide its output.**

A prompt tells the AI model **what to do** and helps guide the model toward the desired output.

## Example

    from langchain_openai import ChatOpenAI
    from dotenv import load_dotenv

    load_dotenv()

    model = ChatOpenAI(
        model="gpt-4",
        temperature=1.5,
        max_completion_tokens=10
    )

    result = model.invoke("Write a 5 line poem on cricket")

    print(result.content)

## Understanding the Example

In the above code:

**"Write a 5 line poem on cricket"** is the **prompt**.

It is passed to the model using:

    model.invoke("Write a 5 line poem on cricket")

The model receives this instruction and generates an output based on it.

## Types of Prompts

### 1. Text-Based Prompts

Traditional prompts are **text-based**.

Example:

> Explain machine learning in simple words.

The input is text and the model generates a response based on that text.

### 2. Multimodal Prompts

Modern AI models can work with multiple types of input, not only text.

A multimodal prompt can contain:

- **Text**
- **Image**
- **Audio / Sound**
- **Video**

Examples:

- Text + Image
- Text + Audio
- Text + Video
- Text + Image + Audio
- Text + Image + Video

## Simple Understanding

### Text-Based Prompt

**Text → Model → Text**

### Multimodal Prompt

**Text / Image / Audio / Video → Model → Output**

## Key Takeaways

- A **prompt** is an input instruction or query given to an AI model.
- Prompts are used to **guide the model's output**.
- A prompt can be a question, instruction, command, or detailed description.
- Traditional prompts are mainly **text-based**.
- Modern multimodal models can accept **text, images, audio, and video** as inputs.
- The quality and clarity of a prompt can strongly affect the usefulness of the model's output.

# Static vs Dynamic Prompts

## Research Assistant

A **static prompt** is a fixed prompt where the instructions and input are written directly into the prompt.

Example:

Summarize the Attention is all you need paper in simple fashion.

In a research assistant application, the user can provide a paper and click **Summarize**. The application then sends the request to the LLM.

## Dynamic Prompt

A **dynamic prompt** is a reusable prompt template where parts of the prompt are replaced with values provided at runtime.

Example:

Please summarize the research paper titled "{paper_input}" with the following specifications:

Explanation Style: {style_input}

Explanation Length: {length_input}

1. Mathematical Details:
   - Include relevant mathematical equations if present in the paper.
   - Explain the mathematical concepts using simple, intuitive code snippets where applicable.

2. Analogies:
   - Use relatable analogies to simplify complex ideas.

If certain information is not available in the paper, respond with:
"Insufficient information available" instead of guessing.

Ensure the summary is clear, accurate, and aligned with the provided style and length.

## Meaning of the Variables

- `{paper_input}` → The research paper provided by the user.
- `{style_input}` → The explanation style selected by the user.
- `{length_input}` → The required explanation length.

## Static vs Dynamic

| Static Prompt | Dynamic Prompt |
|---|---|
| Prompt is fixed | Prompt is reusable |
| Input is hard-coded | Input is provided dynamically |
| Less flexible | More flexible |
| Suitable for simple tasks | Suitable for applications |
| Same instructions every time | Instructions can adapt to user input |

## Static Prompt Example

Summarize the Attention Is All You Need paper in simple fashion.

## Dynamic Prompt Example

Please summarize the research paper titled "{paper_input}" with explanation style "{style_input}" and explanation length "{length_input}".

## Key Idea

A **static prompt** is fixed and does not change based on user input.

A **dynamic prompt** uses variables/placeholders so that the prompt can change according to the user's input.

For example:

Static:
Summarize this research paper in simple language.

Dynamic:
Please summarize the research paper titled "{paper_input}" using "{style_input}" explanation style and "{length_input}" length.

## Why Dynamic Prompts Are Useful

Dynamic prompts are especially useful when building real-world AI applications because the same prompt template can handle:

- Different users
- Different documents
- Different explanation styles
- Different output lengths
- Different tasks
- Different user requirements

## Simple Flow

User Input → Dynamic Prompt Template → LLM → Customized Output

Example:

Paper + Explanation Style + Length
→ Dynamic Prompt
→ LLM
→ Customized Summary

## Important Point

The main difference is:

**Static Prompt = Fixed instructions**

**Dynamic Prompt = Reusable instructions + Variables**

Dynamic prompts make AI applications more flexible, reusable, and user-specific.

# Prompt Template

## Definition

A **PromptTemplate** in LangChain is a structured way to create prompts dynamically by inserting variables into a predefined template.

Instead of hardcoding prompts, `PromptTemplate` allows you to define **placeholders** that can be filled at runtime with different inputs.

This makes prompts:

- Reusable
- Flexible
- Easy to manage
- Suitable for dynamic user inputs
- Useful for automated workflows

## Basic Idea

A PromptTemplate contains:

**Fixed instructions + Variables/Placeholders**

Example:

Please explain {topic} in {language}.

Here:

- `{topic}` → dynamic input
- `{language}` → dynamic input

The values are provided at runtime.

Example inputs:

- topic = "Machine Learning"
- language = "English"

Final prompt:

Please explain Machine Learning in English.

## Why Use PromptTemplate Over f-Strings?

### 1. Default Validation

`PromptTemplate` can validate the variables required by the template.

For example, if the template requires `{topic}` but you forget to provide `topic`, LangChain can identify the missing variable.

### 2. Reusable

The same template can be reused with different inputs.

Example:

Template:

Explain {topic} in simple language.

It can be used for:

- Explain Python in simple language.
- Explain SQL in simple language.
- Explain Deep Learning in simple language.
- Explain Transformers in simple language.

### 3. LangChain Ecosystem

`PromptTemplate` is designed to work naturally with the **LangChain ecosystem**.

It can be combined with:

- LLMs
- Chat models
- Prompt chains
- Output parsers
- LCEL
- Other LangChain components

## Simple Example

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple language."
)

result = prompt.format(topic="Transformers")

print(result)

Output:

Explain Transformers in simple language.

## Key Concept

**f-string:**

A Python string formatting technique.

**PromptTemplate:**

A LangChain component specifically designed for creating reusable and dynamic prompts.

### Remember

`PromptTemplate = Reusable Prompt + Variables + Runtime Inputs`

The main purpose of `PromptTemplate` is to make prompts **dynamic, reusable, structured, and easier to integrate into LangChain applications**.

# Messages in LangChain

## 1. What are Messages?

In LangChain, **messages** represent different types of communication between the user, AI model, and system.

A conversation with an LLM can contain different message roles:

- **SystemMessage** → Gives instructions, rules, behavior, or context to the AI.
- **HumanMessage** → Contains the user's input/question.
- **AIMessage** → Contains the AI/model's previous response.

### Basic Conversation Flow

System Message → Human Message → AI Message → Human Message → AI Message

This structure is useful for **multi-turn conversations**, where the model needs previous interactions as context.

---

## 2. Two Ways to Invoke a Model

A model can be invoked with either:

### A. Single Message

Used for a **single-turn / standalone query**.

Example:

    response = model.invoke("Explain Generative AI")

Here, we provide only one message/query to the model.

Flow:

    Model
      ↓
    Single Message
      ↓
    Response

---

### B. List of Messages

Used for a **multi-turn conversation**.

Example:

    messages = [
        SystemMessage(content="You are a helpful AI assistant."),
        HumanMessage(content="What is LangChain?"),
        AIMessage(content="LangChain is a framework for building LLM applications."),
        HumanMessage(content="Why is it useful?")
    ]

    response = model.invoke(messages)

Here, the model receives the conversation history as a **list of messages**.

Flow:

    Model
      ↓
    List of Messages
      ↓
    Multi-turn Conversation
      ↓
    Response

---

# 3. Types of Messages

## 3.1 SystemMessage

A `SystemMessage` provides **instructions, rules, behavior, or context** to the AI model.

Example:

    from langchain_core.messages import SystemMessage

    message = SystemMessage(
        content="You are a helpful Python tutor."
    )

### Purpose

- Defines AI behavior
- Sets the role/personality
- Provides instructions
- Establishes rules/context

Example:

    System: You are a Python tutor. Explain concepts simply.

---

## 3.2 HumanMessage

A `HumanMessage` represents the **user's input**.

Example:

    from langchain_core.messages import HumanMessage

    message = HumanMessage(
        content="Explain list comprehension in Python."
    )

Example:

    Human: Explain list comprehension in Python.

---

## 3.3 AIMessage

An `AIMessage` represents a **previous response generated by the AI model**.

Example:

    from langchain_core.messages import AIMessage

    message = AIMessage(
        content="List comprehension is a concise way to create lists in Python."
    )

Example:

    AI: List comprehension is a concise way to create lists in Python.

AI messages are especially important when maintaining **conversation history**.

---

# 4. Static Message vs Dynamic Message

Messages can also be divided into:

1. **Static Message**
2. **Dynamic Message**

---

## Static Message

A static message contains fixed text.

Example:

    SystemMessage(
        content="You are a helpful AI assistant."
    )

The content does not change unless we manually change the code.

---

## Dynamic Message

A dynamic message contains variables/placeholders whose values can change at runtime.

Example:

    from langchain_core.prompts import ChatPromptTemplate

    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a helpful {role}."),
        ("human", "Explain {topic}.")
    ])

Values can be supplied dynamically:

    messages = prompt.invoke({
        "role": "Python tutor",
        "topic": "list comprehension"
    })

The final messages become conceptually:

    System: You are a helpful Python tutor.
    Human: Explain list comprehension.

---

# 5. Single Message vs List of Messages

The important distinction is:

    Model
    │
    ├── Single Message
    │      ├── Static Message
    │      └── Dynamic Message
    │
    └── List of Messages
           ├── Static Messages
           │      ├── SystemMessage
           │      ├── HumanMessage
           │      └── AIMessage
           │
           └── Dynamic Messages
                  └── ChatPromptTemplate

### Single Message

Used mainly for:

    Single-turn / standalone queries

Example:

    model.invoke("What is Python?")

### List of Messages

Used mainly for:

    Multi-turn conversations

Example:

    model.invoke([
        SystemMessage(content="You are a teacher."),
        HumanMessage(content="What is Python?"),
        AIMessage(content="Python is a programming language."),
        HumanMessage(content="What is it used for?")
    ])

The model can use the previous messages as conversation context.

---

# 6. ChatPromptTemplate

`ChatPromptTemplate` is used to create **dynamic chat prompts containing multiple messages**.

Example:

    from langchain_core.prompts import ChatPromptTemplate

    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a helpful {role}."),
        ("human", "Explain {topic} in {style}.")
    ])

Then:

    messages = prompt.invoke({
        "role": "teacher",
        "topic": "Generative AI",
        "style": "simple language"
    })

This creates messages dynamically.

### Why ChatPromptTemplate?

It allows us to:

- Create reusable prompts
- Use multiple messages
- Insert dynamic variables
- Build multi-turn/chat-based applications
- Separate prompt structure from actual input values

---

# 7. Message Flow in LangChain

A typical chat application can work like this:

    User Input
        ↓
    ChatPromptTemplate
        ↓
    Dynamic Messages
        ↓
    SystemMessage
    HumanMessage
    AIMessage
    HumanMessage
        ↓
    LLM / Chat Model
        ↓
    AI Response

---

# 8. Complete Example

    from langchain_openai import ChatOpenAI
    from langchain_core.prompts import ChatPromptTemplate

    model = ChatOpenAI(model="gpt-4")

    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a helpful {role}."),
        ("human", "Explain {topic} in simple language.")
    ])

    messages = prompt.invoke({
        "role": "teacher",
        "topic": "LangChain"
    })

    response = model.invoke(messages)

    print(response.content)

Flow:

    System Message
          ↓
    "You are a helpful teacher."
          ↓
    Human Message
          ↓
    "Explain LangChain in simple language."
          ↓
    Chat Model
          ↓
    AI Message / Response

# Final Summary

**Messages are the basic communication units used by LangChain chat models.**

There are three major message types:

1. **SystemMessage** → tells the AI how to behave.
2. **HumanMessage** → represents the user's input.
3. **AIMessage** → represents the AI's response.

A model can receive:

- **One message** → suitable for a standalone query.
- **A list of messages** → suitable for multi-turn conversations.

Messages can be:

- **Static** → fixed content.
- **Dynamic** → content generated using variables.

For dynamic chat applications, `ChatPromptTemplate` is commonly used because it allows us to define reusable message structures with placeholders.

The overall idea is:

    User Input
        ↓
    Prompt / ChatPromptTemplate
        ↓
    Messages
        ↓
    Chat Model
        ↓
    AIMessage / Response

MESSAGE PLACEHOLDER

A MessagesPlaceholder in LangChain is a special placeholder used inside a ChatPromptTemplate to dynamically insert chat history or a list of messages at runtime.
