# Vector Stores in LangChain

# 1. Introduction

A **Vector Store** is a database or storage system designed to store and search **vector embeddings**.

In LangChain, Vector Stores are commonly used in:

- RAG (Retrieval-Augmented Generation)
- Semantic search
- Document retrieval
- Question answering
- Recommendation systems
- Similarity search
- Knowledge-base applications

The basic idea is:

    Document
       ↓
    Text Splitter
       ↓
    Chunks
       ↓
    Embedding Model
       ↓
    Vectors
       ↓
    Vector Store

At query time:

    User Query
       ↓
    Query Embedding
       ↓
    Vector Store
       ↓
    Similar Documents / Chunks
       ↓
    LLM
       ↓
    Answer

---

# 2. What is a Vector?

A **vector** is a numerical representation of data.

For example, text:

    "LangChain is a framework for building LLM applications."

can be converted into a vector such as:

    [0.12, -0.43, 0.87, 0.21, ...]

The actual vector may contain hundreds or thousands of dimensions depending on the embedding model.

The numbers themselves are not usually meaningful to humans.

Their purpose is to represent the semantic characteristics of the input.

---

# 3. What is a Vector Embedding?

An **embedding** is a numerical representation of data that captures useful semantic relationships.

For example:

    "What is LangChain?"

and:

    "Explain the LangChain framework."

have different text but similar meaning.

An embedding model attempts to represent them as vectors that are relatively close in vector space.

Conceptually:

    "What is LangChain?"
            ↓
        Embedding
            ↓
       [0.12, 0.45, ...]


    "Explain LangChain."
            ↓
        Embedding
            ↓
       [0.13, 0.44, ...]

These vectors may have high similarity.

---

# 4. What is a Vector Store?

A **Vector Store** stores vectors and provides mechanisms for searching them.

A vector store typically contains:

    Vector
      +
    Document / Text
      +
    Metadata

Conceptually:

    ┌─────────────────────────────────────┐
    │ Vector Store                        │
    ├─────────────────────────────────────┤
    │ Vector                              │
    │ Document / Chunk                    │
    │ Metadata                            │
    ├─────────────────────────────────────┤
    │ Vector                              │
    │ Document / Chunk                    │
    │ Metadata                            │
    ├─────────────────────────────────────┤
    │ Vector                              │
    │ Document / Chunk                    │
    │ Metadata                            │
    └─────────────────────────────────────┘

The main purpose is:

> Store embeddings and efficiently retrieve vectors that are similar to a query vector.

---

# 5. Why Do We Need Vector Stores?

Traditional databases are generally optimized for exact or structured queries.

For example:

    SELECT * FROM users
    WHERE name = 'John';

But semantic search asks a different question:

> Which documents are conceptually most similar to this query?

Example:

    Query:
    "How can I execute multiple Runnables?"

A keyword search might look for:

    execute
    multiple
    Runnables

A vector search can identify documents discussing:

    RunnableParallel
    parallel execution
    concurrent Runnable operations

even if the exact query words do not appear.

This is the main advantage of semantic vector search.

# 10. Vector Store vs Vector Database

These terms are often used interchangeably, but there can be a conceptual distinction.

### Vector Store

A general abstraction for storing and searching vectors.

### Vector Database

A database system specifically designed to store, index, and search vectors, often with additional production capabilities.

Examples of vector databases include:

- Pinecone
- Qdrant
- Weaviate
- Milvus

LangChain provides integrations that allow applications to work with many vector stores and databases.

# 20. Similarity Search

The core operation of many vector stores is:

> Find vectors that are most similar to the query vector.

Example:

    Query
      ↓
    Embedding
      ↓
    Query Vector
      ↓
    Compare against stored vectors
      ↓
    Find nearest vectors
      ↓
    Return corresponding Documents

---

# 21. Similarity Metrics

Common vector similarity/distance metrics include:

1. Cosine similarity
2. Euclidean distance
3. Dot product

The exact metric depends on:

- Vector database
- Index configuration
- Embedding model
- Application

---

# 22. Cosine Similarity

Cosine similarity measures the angle between two vectors.

Conceptually:

    Vector A
       ↗
      /
     / θ
    ─────────→ Vector B

If the vectors point in similar directions, their cosine similarity is high.

A common interpretation is:

    High similarity → semantically similar
    Low similarity  → less similar

---

# 23. Euclidean Distance

Euclidean distance measures the geometric distance between vectors.

Conceptually:

    A ●
       \
        \
         \ distance
          \
           ● B

Smaller distance means the vectors are closer according to the chosen metric.

---

# 24. Dot Product

The dot product is another similarity measure.

Conceptually:

    A · B

The interpretation depends on how the embedding vectors are generated and normalized.

---

# 25. Similarity vs Distance

Be careful with terminology.

### Similarity

Higher value generally means:

    More similar

### Distance

Lower value generally means:

    More similar

For example:

    Similarity:
    0.95 → very similar

    Distance:
    0.05 → very close

Always check what a particular API returns.

    Top 3 Documents

    Result 1 → Score: 0.94
    Result 2 → Score: 0.89
    Result 3 → Score: 0.84

These documents can then be passed to the LLM.

# 34. Important Embedding Rule

The embeddings stored in the vector store and the query embedding should be compatible.

Ideally:

    Documents
       ↓
    Embedding Model A
       ↓
    Stored Vectors

and:

    Query
       ↓
    Embedding Model A
       ↓
    Query Vector

Using an incompatible embedding space can produce meaningless similarity results.

# 39. Maximum Marginal Relevance (MMR)

**MMR** stands for:

> Maximum Marginal Relevance

MMR attempts to balance:

1. Relevance to the query
2. Diversity among retrieved documents

Why is this useful?

Suppose the top results are:

    Chunk A → very relevant
    Chunk B → almost identical to A
    Chunk C → almost identical to A

Retrieving all three may provide redundant information.

MMR attempts to retrieve:

    Relevant + Diverse

chunks.

---

# 40. Similarity Search vs MMR

### Similarity Search

Focuses mainly on:

    Query ↔ Document similarity

### MMR

Balances:

    Query ↔ Document relevance

and:

    Document ↔ Document diversity

Conceptually:

    Similarity:
    Query → A
    Query → B
    Query → C

    MMR:
    Query → A
    Query → B (different from A)
    Query → C (different from A and B)

# 47. Vector Store and Indexing

Searching millions of vectors by comparing the query against every vector can be expensive.

Vector databases therefore use indexing techniques.

Conceptually:

    Millions of Vectors
           ↓
        Index
           ↓
    Efficient Search
           ↓
       Top Results

Different vector systems use different indexing algorithms.

---

# 48. Approximate Nearest Neighbor Search

Many vector search systems use **Approximate Nearest Neighbor (ANN)** methods.

The goal is to find vectors that are very close to the query without exhaustively comparing against every vector.

Conceptually:

    Query Vector
         ↓
    ANN Index
         ↓
    Candidate Vectors
         ↓
    Top-K Results

This improves search performance at large scale.

---

# 49. Exact vs Approximate Search

### Exact Search

Compares against all relevant vectors.

Advantages:

- Exact result

Disadvantages:

- Can become expensive at large scale

### Approximate Search

Uses indexing techniques to find highly similar vectors efficiently.

Advantages:

- Faster
- Scales better

Disadvantages:

- May trade a small amount of recall for speed

# 61. Simple LangChain Example

A generic vector store workflow can look like:

    from langchain_text_splitters import RecursiveCharacterTextSplitter

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,

    vector_store = SomeVectorStore.from_documents(
        chunks,
        embedding=embedding_model

        "What is LangChain?",
        k=3

# 69. Vector Store vs Document Store

A document store primarily stores documents.

A vector store primarily enables efficient vector similarity search.

A RAG system may use both:

    Original Documents
          ↓
    Document Storage

and:

    Document Chunks
          ↓
    Embeddings
          ↓
    Vector Store

The vector store may also retain the actual chunk text and metadata.

# 70. Common Mistakes

## Mistake 1: Confusing embeddings with vectors

An embedding is the representation produced by an embedding model.

The resulting numerical representation is commonly called a vector.

---

## Mistake 2: Thinking vector stores understand text directly

The vector store generally searches vectors.

The embedding model converts text into vectors.

---

## Mistake 3: Using different embedding models for indexing and querying

This can produce incompatible vector spaces.

Prefer the same compatible embedding model for both stages.

---

## Mistake 4: Assuming more chunks always improve RAG

More chunks can increase:

- Storage
- Embedding cost
- Search complexity
- Duplicate retrieval

Chunk quality matters more than simply increasing chunk count.

---

## Mistake 5: Ignoring metadata

Metadata can be essential for:

- Filtering
- Citations
- Debugging
- Access control
- Document tracking

---

## Mistake 6: Retrieving too many chunks

If you retrieve too many chunks:

    More Context
         ↓
    More Noise
         ↓
    Potentially Worse Answer

Start with a reasonable `k` and evaluate retrieval quality.

---

## Mistake 7: Assuming top similarity means correct answer

A highly similar chunk is not necessarily sufficient to answer the user's question.

Always evaluate:

- Relevance
- Completeness
- Correctness
- Context

# 71. Vector Store Best Practices

## 1. Choose the embedding model carefully

The quality of embeddings directly affects semantic retrieval.

## 2. Design good chunks

Store useful source information.

# 79. Interview Questions

## Q1. What is a Vector Store?

A Vector Store is a storage and search system that stores vector embeddings and allows efficient similarity-based retrieval.

---

## Q2. Why are Vector Stores used in RAG?

They allow the system to retrieve semantically relevant document chunks for a user's query, which can then be provided to an LLM as context.

---

## Q3. What is the difference between an embedding model and a Vector Store?

An embedding model converts text into vectors.

A Vector Store stores and searches those vectors.

---

## Q4. What is similarity search?

Similarity search finds stored vectors that are most similar to a query vector according to a selected similarity or distance metric.

---

## Q5. What is Top-K retrieval?

Top-K retrieval returns the K most relevant results.

For example:

    k = 5

means the system attempts to return five relevant results.

---

## Q6. What is MMR?

MMR stands for Maximum Marginal Relevance. It attempts to balance relevance to the query with diversity among retrieved documents.

---

## Q7. What is metadata filtering?

Metadata filtering restricts retrieval based on metadata such as:

    source
    category
    date
    document_id

---

## Q8. What is a Retriever?

A Retriever is an abstraction that takes a query and returns relevant Documents.

A vector store can be converted into a retriever.

---

## Q9. Can a Vector Store generate answers?

No. A Vector Store retrieves information. The LLM is responsible for generating the natural-language answer.

---

## Q10. Why is the embedding model important?

Because it determines how text is represented in vector space. Poor embeddings can result in poor semantic retrieval.

---

## Q11. What is hybrid search?

Hybrid search combines semantic/vector search with lexical or keyword search.

---

## Q12. What is ANN?

ANN stands for Approximate Nearest Neighbor. It is a family of techniques used to efficiently find vectors close to a query vector without exhaustively comparing every vector.
