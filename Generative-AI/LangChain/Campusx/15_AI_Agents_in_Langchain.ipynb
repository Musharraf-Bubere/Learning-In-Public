# AI Agents in LangChain

## 1. What is an AI Agent?

An **AI Agent** is an AI system that can understand a user's goal, decide what actions are required, use available tools, observe the results, and continue taking actions until the task is completed.

A normal LLM mainly follows:

    User
      ↓
     LLM
      ↓
    Answer

An AI Agent can follow:

    User
      ↓
    Agent
      ↓
    Think / Decide
      ↓
    Select Tool
      ↓
    Execute Tool
      ↓
    Observe Result
      ↓
    Decide Next Action
      ↓
    Tool / Answer
      ↓
    Final Answer

### Simple Definition

> An AI Agent is an LLM-powered system that dynamically decides what actions or tools to use to accomplish a goal.

---

# 2. Why Do We Need AI Agents?

LLMs are excellent at language understanding and generation, but many real-world tasks require multiple actions.

Example:

    User:
    "Find the weather in Mumbai, calculate the temperature
    in Fahrenheit, and explain whether I should carry an umbrella."

The system may need to:

    1. Get weather information.
    2. Perform a temperature conversion.
    3. Analyze the weather information.
    4. Generate a final response.

A fixed workflow could do this, but an agent can dynamically determine what steps are necessary.

      ↓

      ↓

      ↓
    Agent
      ↓

# 4. Agentic AI

**Agentic AI** refers to AI systems capable of pursuing goals through dynamic decision-making and actions.

A simplified agentic loop is:

    Goal
      ↓
    Understand
      ↓
    Plan / Decide
      ↓
    Act
      ↓
    Observe
      ↓
    Decide Again
      ↓
    Act
      ↓
    ...
      ↓
    Goal Completed

LangChain provides abstractions for building such systems.

# 9. Agent Control Loop

The most important concept is the **agent loop**.

    User Query
        ↓
       Agent
        ↓
    LLM decides
        ↓
    Tool required?
       / \
     Yes  No
      ↓    ↓
    Tool  Final Answer
      ↓
    Result
      ↓
     Agent
      ↓
    LLM decides again
      ↓
    Tool / Answer

The loop continues until the agent determines that it has enough information to provide the final response.

---

# 10. Agent Reasoning

At a high level, an agent reasons about:

    What is the user asking?

    What information do I need?

    Which tool can provide it?

    What arguments should I pass?

    What did the tool return?

    Do I need another tool?

    Is the task complete?

The internal reasoning process of modern models should not be confused with exposing a hidden chain-of-thought.

The important application-level concept is the observable action loop:

    Decide → Act → Observe → Continue

# 19. Tool Calling in Agents

Tool calling is the mechanism that allows the model to request tool execution.

Example:

    User:
    "Calculate 20 × 5."

    LLM:
    Tool Call:
        calculator
        expression = "20 * 5"

    Runtime:
        executes calculator

    Tool:
        100

    LLM:
        "The answer is 100."

The agent provides the larger decision loop around these tool calls.

# 20. Agent Loop vs Tool Calling

Important distinction:

### Tool Calling

    LLM
      ↓
    Tool Call
      ↓
    Tool
      ↓
    Result

### Agent

    LLM
      ↓
    Tool Call
      ↓
    Tool
      ↓
    Result
      ↓
    LLM
      ↓
    Another Tool Call
      ↓
    Tool
      ↓
    Result
      ↓
    Final Answer

Therefore:

> Tool calling is a capability; agent orchestration is the control loop that can use that capability repeatedly.

# 26. Agent State vs Memory

These concepts are related but should not be treated as identical.

### State

Information needed by the current workflow/execution.

### Memory

Information retained across interactions or sessions.

For example:

    Current execution:
    User asked about order #123
    → State

    Previous conversations:
    User's saved preferences
    → Memory

Modern LangChain agent architectures can use persistent state/checkpointing mechanisms when long-running or multi-turn behavior is required.

---

# 27. Stateless Agent

A stateless agent does not retain previous interaction state after the execution ends.

Example:

    User → Agent → Answer

New request:

    User → Agent → Answer

The second request may not know the first request unless relevant state is passed again.

---

# 28. Stateful Agent

A stateful agent can preserve relevant state.

Example:

    User:
    "Find my order."

    Agent:
    "Your order is #123."

    User:
    "Cancel it."

The agent can use previous context to understand:

    "it" = order #123

State is especially important for multi-turn workflows.

---

# 29. Short-Term vs Long-Term Memory

### Short-Term Memory

Conversation/workflow state during an interaction.

Example:

    User:
    "My order number is 123."

    Later:
    "What's its status?"

### Long-Term Memory

Information intentionally stored for future interactions.

Example:

    User preference:
    "I prefer concise responses."

Long-term memory requires deliberate storage and privacy/security considerations.

# 37. Agent Termination

An agent needs a stopping condition.

Typical termination:

    Agent determines task is complete
              ↓
         Final Response

Other possible termination conditions:

- Maximum iterations reached
- Maximum execution time reached
- Tool error
- Human intervention
- Safety policy triggered

Without proper limits, an agent can loop unnecessarily.

# 43. Read vs Write Tools

Agents should treat these differently.

### Read Tool

    search_docs()
    get_order()
    get_weather()

Usually lower risk.

### Write Tool

    delete_order()
    send_email()
    transfer_money()
    update_database()

Higher risk.

Write operations should generally have stronger authorization and validation.

# 44. Agent Security

Agent security is critical because agents can combine:

    LLM reasoning
       +
    External tools
       +
    Real-world actions

Risks include:

### Prompt Injection

Malicious instructions attempt to manipulate the agent.

### Tool Abuse

The model is induced to call a dangerous tool.

### Data Leakage

Sensitive information is exposed.

### Excessive Permissions

Agent can access more systems than necessary.

### Unauthorized Actions

Agent performs an action the user is not allowed to perform.

### Malicious Tool Output

An external system returns content designed to influence the agent.

---

# 45. Least Privilege for Agents

Only provide tools that the agent needs.

Bad:

    Agent
      ├── delete_database
      ├── transfer_money
      ├── send_email
      ├── execute_code
      ├── modify_users
      └── search_docs

If the agent only needs document search:

    Agent
      └── search_docs

This reduces risk.

---

# 46. Tool Authorization

The agent should not be the final authority.

Example:

    Agent:
    "Call transfer_money(amount=50000)."

The tool layer should still verify:

    Is user authenticated?
    Is user authorized?
    Is account valid?
    Is transaction permitted?
    Is amount within limits?

Only then:

    Execute transaction

# 53. Agent vs Chain vs Workflow

| Feature | Chain | Workflow | Agent |
|---|---|---|---|
| Fixed sequence | Usually | Yes | No |
| Dynamic decisions | Limited | Limited/controlled | Strong |
| Tool use | Possible | Possible | Central capability |
| Predictability | High | High | Lower |
| Flexibility | Medium | High | High |
| Debugging | Easier | Easier | More complex |
| Cost control | Easier | Easier | More challenging |
| Best for | Simple pipelines | Known processes | Open-ended tasks |

# 54. When Should You Use an Agent?

Use an agent when:

- The next step depends on previous results.
- Multiple tools may be required.
- The required sequence is not known beforehand.
- The user asks open-ended tasks.
- Dynamic decision-making is valuable.
- The system must interact with several external systems.

Example:

    "Research this company, compare its products,
     calculate pricing differences, and summarize
     the findings."

This may require:

    Search
    Website retrieval
    Calculator
    Document analysis

An agent can coordinate these actions.

---

# 55. When Should You NOT Use an Agent?

Do not use an agent when the workflow is completely predictable.

Example:

    User
      ↓
    Load PDF
      ↓
    Split
      ↓
    Embed
      ↓
    Store

There is no need for an agent to decide these steps.

A deterministic pipeline is better.

Another example:

    Input
      ↓
    Prompt
      ↓
    LLM
      ↓
    Parser
      ↓
    Output

A chain is sufficient.

---

# 56. Agent Design Principle

A good rule:

    Known process
       ↓
    Workflow / Chain

    Unknown or dynamic process
       ↓
    Agent

This reduces unnecessary complexity.

# 58. Agent Evaluation

Evaluate more than just the final answer.

Important metrics:

### Task Success

Did the agent accomplish the user's goal?

### Tool Selection Accuracy

Did it choose the correct tools?

### Tool Argument Accuracy

### Efficiency

How many tool calls were required?

### Latency

How long did execution take?

### Cost

How many model/tool resources were consumed?

### Safety

# 59. Agent Failure Modes

## Failure 1: Wrong Tool

Agent selects an inappropriate tool.

Solution:

    Better descriptions
    Better schemas
    Tool routing
    Evaluation

---

## Failure 2: Wrong Arguments

Example:

    get_order(order_id="abc")

when the expected ID is numeric.

Solution:

    Structured schemas
    Validation

---

## Failure 3: Tool Loop

    Agent
      ↓
    Tool
      ↓
    Agent
      ↓
    Same Tool
      ↓
    Agent
      ↓
    ...

Solution:

    Iteration limits
    Better termination logic
    Error handling

---

## Failure 4: Hallucinated Tool Result

The model acts as though a tool returned something it did not.

Solution:

    Use actual tool messages/results
    Keep tool execution separate
    Validate state

---

## Failure 5: Excessive Tool Calls

The agent repeatedly calls tools unnecessarily.

Solution:

    Better prompts
    Better tool descriptions
    Limits
    Caching
    Deterministic workflows where appropriate

---

## Failure 6: Unauthorized Action

The model calls a tool the user should not be able to use.

Solution:

    Application-level authorization
    Least privilege
    Human approval

    LLM call
      +
    Tool call
      +
    LLM call

Therefore:

    More steps
      ↓
    More latency
      +
    More model/tool cost

Agents should be designed for efficiency.

Optimization techniques include:

- Reduce unnecessary tool calls

- Cache repeated results
- Use appropriate models
- Use deterministic workflows for predictable tasks

# 70. Agent Tool Design

Good tools should be:

### Specific

    get_order()

instead of:

    do_anything()

### Clearly Described

### Safe

Validate input and authorization.

# 71. Agent System Design Checklist

Before building an agent, ask:

    [ ] Is an agent actually required?
    [ ] Could a deterministic workflow solve this?
    [ ] What tools are required?
    [ ] Are tool descriptions clear?
    [ ] Are tool schemas structured?
    [ ] Are tool inputs validated?
    [ ] Are permissions enforced?
    [ ] Are write actions protected?
    [ ] Is human approval required?
    [ ] What is the termination condition?
    [ ] What is the maximum iteration limit?
    [ ] What happens when a tool fails?
    [ ] How will tool calls be monitored?
    [ ] How will task success be evaluated?
    [ ] How will cost and latency be controlled?

# 72. Common Beginner Mistakes

### Mistake 1

Thinking an agent is just an LLM.

    Agent ≠ LLM

An agent is an orchestration system using an LLM plus tools/state/control logic.

---

### Mistake 2

Thinking agents always need dozens of tools.

They do not.

A simple agent may need only:

    LLM + 1 or 2 tools

---

### Mistake 3

Using an agent for every problem.

If the workflow is predictable:

    Use a chain/workflow.

---

### Mistake 4

Trusting the agent with unrestricted permissions.

Never do this for sensitive systems.

---

### Mistake 5

Ignoring termination.

Agents need limits and stopping conditions.

---

### Mistake 6

Ignoring observability.

Without traces, debugging complex agent behavior becomes difficult.

---

### Mistake 7

Assuming agents are always better.

Agents provide flexibility at the cost of:

    Complexity
    Latency
    Cost
    Less predictability

# 80. Interview Questions

## Beginner

### Q1. What is an AI Agent?

An AI Agent is an LLM-powered system that can dynamically decide actions, use tools, observe results, and continue until a task is completed.

### Q2. What are the main components of an agent?

    LLM
    Tools
    Agent/control logic
    State/context
    Tool execution

### Q3. What is the difference between an LLM and an agent?

An LLM primarily understands and generates language. An agent adds orchestration, tools, and a decision/action loop around the model.

### Q4. What is a tool?

A tool is an external capability such as a function, API, database query, search engine, or business operation that an agent can invoke.

### Q5. Can an agent use multiple tools?

Yes.

---

# 81. Intermediate Interview Questions

### Q6. What is the difference between an agent and a chain?

A chain generally follows a predefined sequence, while an agent dynamically determines which actions to take.

### Q7. What is tool calling?

Tool calling allows an LLM to request execution of a specific tool with structured arguments.

### Q8. Does an LLM execute the tool?

Usually, the model generates the tool-call request and the application/runtime executes the actual tool.

### Q9. Can RAG be used inside an agent?

Yes. A RAG retriever can be exposed as a tool.

### Q10. Why might you avoid using an agent?

Because agents add:

    Complexity
    Cost
    Latency
    Less predictability

If the workflow is fixed, a deterministic workflow may be better.

    Error handling

Never allow the LLM alone to authorize the transaction.
