# Model


### What are Models?

The Model Component in LangChain is a crucial part of the framework, designed to facilitate interactions with various language models and embedding models.

It abstracts the complexity of working directly with different LLMs, chat models, and embedding models, providing a uniform interface to communicate with them. This makes it easier to build applications that rely on AI-generated text, text embeddings for similarity search, and retrieval-augmented generation (RAG).

# Models in Generative AI

## 1. What is a Model?

A **model** is a computational system that learns patterns from data and uses those learned patterns to perform a specific task.

In Generative AI, models are mainly used to:

- Understand input data
- Learn patterns and relationships
- Generate new content
- Convert information into useful representations
- Perform specific tasks

---

## 3. Language Models

A **Language Model** is a model designed to understand and generate human language.

It learns relationships between words, tokens, sentences, and larger pieces of text.

### Example

```text
Input:
"The capital of France is"

Prediction:
"Paris"
```

At a fundamental level, a language model learns patterns in language and uses the given context to predict or generate appropriate text.

### Applications

- Text generation
- Text completion
- Question answering
- Summarization
- Translation
- Classification
- Information extraction
- Code generation

---

## 4. LLMs — Large Language Models

**LLM = Large Language Model**

An LLM is a large-scale language model trained on very large datasets and typically containing a very large number of parameters.

### What makes an LLM "large"?

The term **large** generally refers to factors such as:

- Large training datasets
- Large number of parameters
- Large computational requirements
- Ability to handle complex language tasks

### Examples

- GPT
- Llama
- Claude
- Gemini
- Mistral

### Applications

- Text generation
- Question answering
- Summarization
- Reasoning
- Code generation
- Content generation
- Information extraction

---

## 5. Chat Models

A **Chat Model** is a language model optimized for **conversational interaction**.

Chat models are designed to work with conversational inputs such as:

- System instructions
- User messages
- Conversation history
- Assistant responses

### Example

```text
User:
What is Python?

Assistant:
Python is a high-level programming language...
```

The model uses the conversation context to generate an appropriate response.

### Important Distinction

**LLM and Chat Model are related but are not exactly the same concept.**

- **LLM:** A broad category of large language models.
- **Chat Model:** A language model optimized or instructed for conversational interaction.

---

## 6. Embedding Models

An **Embedding Model** converts data such as text into a numerical representation called an **embedding**.

An embedding is usually represented as a vector of numbers.

### Example

```text
"Python is a programming language"
                ↓
        Embedding Model
                ↓
[0.21, -0.45, 0.73, 0.11, ...]
```

The vector represents semantic information about the input.

Texts with similar meanings generally have embeddings that are closer to each other in the embedding space.

### Applications

- Semantic search
- Similarity search
- Document retrieval
- Recommendation systems
- Clustering
- Retrieval-Augmented Generation (RAG)

---

## 7. Language Model vs Embedding Model

| Feature | Language Model | Embedding Model |
|---|---|---|
| Main purpose | Understand/generate language | Convert data into vectors |
| Output | Text/tokens | Numerical vector |
| Text generation | Yes | No |
| Semantic search | Indirectly | Yes |
| RAG | Generation part | Retrieval part |
| Main use | Generation and language tasks | Search, similarity and retrieval |

---

## 8. How Models Work Together in RAG

A typical RAG application can use both an **Embedding Model** and an **LLM/Chat Model**.

```mermaid
flowchart LR
    A[User Question] --> B[Embedding Model]
    B --> C[Vector Search]
    C --> D[Relevant Documents]
    D --> E[LLM / Chat Model]
    A --> E
    E --> F[Final Answer]
```

### RAG Flow

1. The user asks a question.
2. The question is converted into an embedding.
3. The embedding is used for vector search.
4. Relevant documents are retrieved.
5. The question and retrieved documents are provided to an LLM/Chat Model.
6. The LLM/Chat Model generates the final answer.

# Language Models

## 1. What are Language Models?

**Language Models (LMs)** are AI systems designed to **process, understand, and generate natural language text**.

They take language-related input and produce language-related output.

### Basic Flow

```text
Text Input
    ↓
Language Model
    ↓
Text Output
```

Language Models are a major component of **Generative AI** and are used for tasks such as text generation, question answering, summarization, translation, coding, and conversation.

---

## 2. Types of Language Models

Language Models can broadly be divided into:

```mermaid
flowchart TD
    A[Language Models] --> B[LLMs]
    A --> C[Chat Models]
```

- **LLMs (Large Language Models)**
- **Chat Models**

---

# 3. LLMs — Large Language Models

**LLM = Large Language Model**

LLMs are **general-purpose language models** that can process and generate text.

Traditionally, an LLM could take a **string/plain text** as input and return a **string/plain text** as output.

### Basic Flow

```text
String / Plain Text
        ↓
       LLM
        ↓
String / Plain Text
```

### Example

```text
Input:
"Explain Python in one sentence:"

        ↓

LLM

        ↓

Output:
"Python is a high-level programming language
used for many types of software development."
```

### Characteristics of LLMs

- Work primarily with text
- Can generate text based on the given context
- General-purpose in nature
- Can perform many language-related tasks
- Can be used for text completion and generation
- Form the foundation of many modern AI applications

### Common Applications

- Text generation
- Text completion
- Summarization
- Question answering
- Translation
- Code generation
- Information extraction
- Content generation

---

# 4. Chat Models

**Chat Models** are language models **specialized for conversational tasks**.

Unlike traditional text-based LLM interfaces, Chat Models work with a **sequence of messages** and return **chat messages**.

### Basic Flow

```text
Sequence of Messages
          ↓
      Chat Model
          ↓
    Chat Message
```

### Example

```text
User:
What is Python?

        ↓

Chat Model

        ↓

Assistant:
Python is a high-level programming language
known for its simplicity and readability.
```

---

## 5. Messages in Chat Models

Chat Models generally work with different types of conversational messages.

### Common Message Types

- **System Message** → Defines instructions or behavior
- **Human/User Message** → Contains the user's input
- **AI/Assistant Message** → Contains the model's response

### Example

```text
System:
You are a helpful Python tutor.

User:
Explain variables in Python.

Assistant:
A variable is a name used to store a value.
```

This message-based structure makes Chat Models suitable for conversational applications.

---

# 6. LLM vs Chat Model

| Feature | LLM | Chat Model |
|---|---|---|
| Main purpose | General text generation | Conversational interaction |
| Input | String / plain text | Sequence of messages |
| Output | String / plain text | Chat message |
| Interaction style | Text completion/generation | Conversation |
| Context | Text-based context | Conversation/message context |
| Typical use | Raw text generation | Chatbots and assistants |

---

# 7. Key Concept

The important difference is **how the model is interfaced**.

```text
LLM

String
  ↓
LLM
  ↓
String
```

Whereas:

```text
Chat Model

Messages
   ↓
Chat Model
   ↓
Messages
```

---

# 8. LLMs vs Chat Models in Modern Applications

Traditionally:

- **LLMs** were commonly used for raw text generation.
- **Chat Models** were designed specifically for conversational interactions.

Modern language models can often support both types of interactions through different interfaces or APIs.

Therefore, the distinction is often more about the **input/output interface and intended interaction style** than completely different underlying model architectures.

# LLMs vs Chat Models

## 1. Overview

Both **LLMs (Large Language Models)** and **Chat Models** are language models, but they are designed for different interaction styles.

### LLMs

LLMs are traditionally designed for **general-purpose, free-form text generation**.

### Chat Models

Chat Models are **instruction-tuned language models** optimized for **conversational interactions**, especially multi-turn conversations.

---

## 2. Basic Difference

```text
LLM
↓
Plain Text Input
↓
Model
↓
Plain Text Output
```

```text
Chat Model
↓
Sequence of Messages
↓
Instruction-Tuned Model
↓
Chat Message Output
```

---

## 3. LLMs — Base Models

A traditional LLM or **base model** is primarily trained on large amounts of general text such as:

- Books
- Articles
- Websites
- Documents
- Other text corpora

Its main objective is generally to learn language patterns and predict/generate text.

### Purpose

**Free-form text generation**

### Example

```text
Input:
"The future of Artificial Intelligence is"

        ↓

LLM

        ↓

Output:
"The future of Artificial Intelligence is likely
to transform many areas of technology and society..."
```

The output is essentially generated as plain text.

---

# 4. Chat Models — Instruction-Tuned Models

Chat Models are language models that have been **fine-tuned or instruction-tuned for conversational tasks**.

They are designed to understand structured conversations involving different roles.

### Common Roles

- **System** → Defines the model's behavior or instructions
- **User** → Provides the user's request
- **Assistant** → Represents the model's response

### Example

```text
System:
You are a helpful Python tutor.

User:
Explain Python lists.

Assistant:
A Python list is an ordered, mutable collection
used to store multiple values.
```

This structure makes Chat Models suitable for conversational AI applications.

---

# 5. LLMs vs Chat Models

| Feature | LLMs (Base Models) | Chat Models (Instruction-Tuned) |
|---|---|---|
| **Purpose** | Free-form text generation | Optimized for multi-turn conversations |
| **Training Data** | General text corpora such as books and articles | Fine-tuned on chat/instruction datasets |
| **Input** | String / plain text | Sequence of messages |
| **Output** | String / plain text | Chat messages |
| **Memory & Context** | No built-in conversational memory | Supports structured conversation history |
| **Role Awareness** | Does not inherently use system/user/assistant roles | Understands system, user, and assistant roles |
| **Interaction** | Text completion/generation | Conversational interaction |
| **Typical Use** | Text generation, summarization, translation, creative writing, code generation | Chatbots, virtual assistants, customer support, AI tutors |
| **Optimization** | General language modeling | Following instructions and maintaining conversations |

---

# 6. Training Difference

The major difference is not necessarily that Chat Models use a completely different architecture.

A common workflow is:

```mermaid
flowchart TD
    A[Large Text Dataset] --> B[Pretraining]
    B --> C[Base Language Model / LLM]

    C --> D[Instruction / Chat Fine-Tuning]
    D --> E[Chat Model]
```

### Base LLM

The model is first trained on large amounts of text to learn:

- Language patterns
- Grammar
- Context
- Relationships between words/tokens
- General knowledge patterns

### Chat Model

The base model can then be further trained or aligned using instruction and conversational data so that it becomes better at:

- Following instructions
- Answering questions
- Maintaining conversational context
- Understanding roles
- Producing helpful responses

---

# 7. Memory & Context

### Traditional LLM

A traditional base LLM does not inherently have persistent conversational memory.

For example:

```text
Request 1:
My name is Alex.

Request 2:
What is my name?
```

Without providing the previous context again, the model does not automatically have access to the previous request.

### Chat Model

A Chat Model can receive conversation history as a structured sequence of messages.

```text
System:
You are a helpful assistant.

User:
My name is Alex.

Assistant:
Nice to meet you, Alex!

User:
What is my name?
```

The conversation history can be provided to the model so it can use that context when generating the next response.

> **Important:** Chat Models do not necessarily have permanent memory by themselves. Conversation history or external memory systems may be provided to them.

---

# 8. Role Awareness

Traditional text-based LLM interaction can look like:

```text
Plain Text
    ↓
"Explain Python"
    ↓
LLM
    ↓
Plain Text
```

Chat Models use structured roles:

```text
System Message
       ↓
User Message
       ↓
Assistant Message
       ↓
Chat Model
```

### Example

```text
System:
You are an expert Python teacher.

User:
Explain loops in Python.

Assistant:
A loop is used to repeatedly execute a block of code...
```

The model can interpret the purpose of each message based on its role.

# Temperature in Language Models

## 1. What is Temperature?

**Temperature** is a parameter that controls the **randomness of a language model's output**.

It affects whether the model produces responses that are:

- More **deterministic and predictable**
- More **random, creative, and diverse**

In simple terms:

> **Lower temperature → More predictable output**

> **Higher temperature → More creative and random output**

---

## 2. How Temperature Works

When a language model generates text, it usually calculates probabilities for possible next tokens.

For example:

```text
"The sky is"

blue      → 0.70
clear     → 0.20
beautiful → 0.08
green     → 0.02
```

The model uses these probabilities to select the next token.

Temperature changes how strongly the model favors high-probability tokens.

```mermaid
flowchart LR
    A[Input Prompt] --> B[Language Model]
    B --> C[Token Probabilities]
    C --> D[Temperature]
    D --> E[Next Token]
```

---

# 3. Low Temperature

A **low temperature** makes the model more likely to choose the highest-probability tokens.

### Result

- More deterministic
- More predictable
- More consistent
- Less creative
- Less variation between responses

### Example

```text
Temperature = 0.0

Prompt:
"The capital of France is"

Likely output:
"Paris"
```

For tasks where correctness and consistency are important, a lower temperature is generally preferred.

### Suitable for

- Mathematical answers
- Factual questions
- Code generation
- Data extraction
- Structured outputs
- Classification

---

# 4. High Temperature

A **higher temperature** makes the probability distribution more spread out, allowing lower-probability tokens to be selected more often.

### Result

- More random
- More creative
- More diverse
- Less predictable
- Potentially less consistent

### Example

```text
Temperature = 1.0+

Prompt:
"Write a story about a robot."

Possible outputs can vary significantly between runs.
```

Higher temperature can be useful when you want multiple creative possibilities.

### Suitable for

- Creative writing
- Storytelling
- Brainstorming
- Generating ideas
- Poetry
- Creative content

---

# 5. Temperature Scale

A common conceptual scale is:

```text
Low Temperature                         High Temperature
      │                                         │
      ▼                                         ▼
  0.0 ───── 0.3 ───── 0.5 ───── 0.7 ───── 1.0 ───── 1.5+
   │           │         │          │          │
Predictable   Stable    Balanced   Creative   Random
```

> **Note:** The exact temperature range and behavior can vary between models and APIs. Some providers may restrict the supported range or use different defaults.

---

# 6. Recommended Temperature by Use Case

| Use Case | Typical Temperature | Goal |
|---|---:|---|
| Factual answers | `0.0 – 0.3` | Accuracy & consistency |
| Mathematics | `0.0 – 0.3` | Deterministic reasoning |
| Code generation | `0.0 – 0.3` | Consistent output |
| General Q&A | `0.5 – 0.7` | Balanced responses |
| Explanations | `0.5 – 0.7` | Natural but controlled |
| Creative writing | `0.9 – 1.2` | Creativity |
| Storytelling | `0.9 – 1.2` | Variety |
| Brainstorming | `1.0+` | Diverse ideas |

These are **general guidelines**, not universal settings.

---

# 7. Temperature and Probability

Temperature is commonly described using the following transformation of model logits:

$$
P_i = \frac{e^{z_i/T}}{\sum_j e^{z_j/T}}
$$

Where:

- $P_i$ = probability of token $i$
- $z_i$ = model's logit for token $i$
- $T$ = temperature
- $e$ = Euler's number

### Effect of Temperature

### When $T$ is low

The probability distribution becomes more concentrated around high-probability tokens.

```text
Token A → 0.90
Token B → 0.07
Token C → 0.02
Token D → 0.01
```

The model is more likely to choose the most probable token.

### When $T$ is high

The distribution becomes flatter.

```text
Token A → 0.45
Token B → 0.25
Token C → 0.20
Token D → 0.10
```

Lower-probability tokens have a greater chance of being selected.

---

# 8. Temperature Does NOT Mean "Intelligence"

Temperature does **not** directly control:

- Model intelligence
- Model knowledge
- Number of parameters
- Training quality
- Reasoning ability

Instead, it controls the **randomness/diversity of token selection** during generation.

```text
Temperature
     ↓
Token Selection Behavior
     ↓
Output Randomness / Diversity
```

It does **not** mean:

```text
Higher Temperature
       ↓
Smarter Model ❌
```

---

Consider the prompt:

```text
"Give me a name for an AI startup."
```

### Low Temperature

```text
Possible output:
"Nova AI"
```

Running the same prompt repeatedly may produce similar results.

### Higher Temperature

```text
Possible outputs:

"NeuraForge"
"MindSpark AI"
"CortexFlow"
"NovaSphere"
"Synaptiq"
```

The outputs can be more varied and creative.

# 10. Temperature in Generative AI Applications

Temperature can be adjusted depending on the requirements of the application.

```mermaid
flowchart TD
    A[Application] --> B{Required Output}

    B -->|Accuracy / Consistency| C[Low Temperature]
    B -->|Balanced| D[Medium Temperature]
    B -->|Creativity / Diversity| E[High Temperature]

    C --> F[Facts / Math / Code]
    D --> G[General Q&A / Explanations]
    E --> H[Stories / Ideas / Brainstorming]
```

---

# 11. Important Practical Point

Temperature is **not a universal guarantee of deterministic behavior**.

Even with a low temperature, exact output behavior can depend on:

- The model
- The API/provider
- Sampling configuration
- Other generation parameters
- System implementation

Therefore, temperature should be treated as a **generation-control parameter**, not a guarantee of exact reproducibility.

- Temperature does **not** increase or decrease the model's intelligence.
- Use lower values for tasks requiring consistency.
- Use higher values for creative tasks.

- Temperature affects **token selection probabilities**.

# Language Models

## LLMs vs Chat Models

### LLMs (Base Models)

Language models that take text as input and return a string (plain text). These are traditionally older models and are not used much now.

### Chat Models (Instruction-Tuned)

Language models specialized for conversational tasks. They take a sequence of messages as inputs and return chat messages as outputs (as opposed to using plain text). These are traditionally newer models and are used more in comparison to traditional LLMs.

| Feature | LLMs (Base Models) | Chat Models (Instruction-Tuned) |
|---|---|---|
| **Purpose** | Free-form text generation | Optimized for multi-turn conversations |
| **Training Data** | General text corpora (books, articles) | Fine-tuned on chat datasets (dialogues, user-assistant conversations) |
| **Memory & Context** | No built-in memory | Supports structured conversation history |
| **Role Awareness** | No understanding of "user" and "assistant" roles | Understands "system", "user", and "assistant" roles |
| **Example Models** | GPT-3, Llama-2-7B, Mistral-7B, OPT-1.3B | GPT-4, GPT-3.5-turbo, Llama-2-Chat, Mistral-Instruct, Claude |
| **Use Cases** | Text generation, summarization, translation, creative writing, code generation | Conversational AI, chatbots, virtual assistants, customer support, AI tutors |

---

# Temperature

`temperature` is a parameter that **controls the randomness of a language model's output**. It affects how creative or deterministic the responses are.

- **Lower values (0.0 - 0.3)** → More deterministic and predictable.
- **Higher values (0.7 - 1.5)** → More random, creative, and diverse.

| Use Case | Recommended Temperature |
|---|---|
| Factual answers (math, code, facts) | `0.0 - 0.3` |
| Balanced response (general QA, explanations) | `0.5 - 0.7` |
| Creative writing, storytelling, jokes | `0.9 - 1.2` |
| Maximum randomness (wild ideas, brainstorming) | `1.5+` |

### Simple Understanding

- **Low temperature** → Model is more predictable and consistent.
- **High temperature** → Model explores more possible/random outputs.

> **Important:** Temperature controls the randomness of token selection. It does not make the model more intelligent.

# Open-Source Models

## Definition

**Open-source language models** are freely available AI models that can be **downloaded, modified, fine-tuned, and deployed** without restrictions from a central provider.

Unlike closed-source models such as OpenAI's GPT-4, Anthropic's Claude, or Google's Gemini, open-source models allow greater control and customization.

---

## Open-Source vs Closed-Source Models

| Feature | Open-Source Models | Closed-Source Models |
|---|---|---|
| **Cost** | Free to use (no API costs) | Paid API usage (e.g., OpenAI charges per token) |
| **Control** | Can modify, fine-tune, and deploy anywhere | Locked to provider's infrastructure |
| **Data Privacy** | Runs locally (no data sent to external servers) | Sends queries to provider's servers |
| **Customization** | Can fine-tune on specific datasets | No access to fine-tuning in most cases |

# Ways to Use Open-Source Models

Open-source models can generally be used in two main ways:

**Open-Source Models**

→ **Using Hugging Face (HF) Inference API**

→ **Running Locally**

## 1. Using Hugging Face (HF) Inference API

Instead of downloading and running the model yourself, you can access an open-source model through an inference API.

### Advantages

- No need to manage GPUs yourself.
- Easier setup.
- Model runs on remote infrastructure.
- Can access models through an API.

### Disadvantages

- May have API/inference costs.
- Requires internet connection.
- Data is sent to an external service.

## 2. Running Locally

You can download the model and run it directly on your own computer/server.

### Advantages

- Greater control.
- Can work without sending data to external servers.
- Better privacy.
- Can customize and fine-tune the model.
- No per-request API cost.

### Disadvantages

- Requires suitable hardware.
- Large models require powerful GPUs/RAM.
- Setup can be complex.

# Disadvantages of Open-Source Models

| Disadvantage | Details |
|---|---|
| **High Hardware Requirements** | Running large models (e.g., LLaMA-2-70B) requires expensive GPUs. |
| **Setup Complexity** | Requires installation of dependencies like PyTorch, CUDA, and Transformers. |
| **Lack of RLHF** | Most open-source models don't have fine-tuning with human feedback, making them weaker in instruction-following. |
| **Limited Multimodal Abilities** | Open models may have limited support for images, audio, or video compared with some proprietary multimodal models. |

# Key Takeaways

- **LLMs** are base language models primarily designed for text completion/generation.
- **Chat Models** are optimized for conversations and understand message roles such as `system`, `user`, and `assistant`.
- **Temperature** controls the randomness of model output.
- **Low temperature** → more deterministic/predictable output.
- **High temperature** → more random/creative output.
- **Open-source models** can generally be downloaded, modified, fine-tuned, and deployed with greater flexibility.
- **Hugging Face** is a major platform/repository for finding and using open-source models.
- Open-source models can be used through an **Inference API** or **run locally**.
- Running models locally provides greater control and privacy but may require significant hardware and setup.
- Large models require more **GPU memory, RAM, storage, and computational resources**.
- Open-source does not always mean completely unrestricted or zero-cost; model licenses and infrastructure costs can still apply.
