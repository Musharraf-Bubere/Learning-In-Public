# Chromadb

Chromadb is a AI-native open-source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0.

```python
# building a sample vectordb
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("speech.txt")
data = loader.load()
data
```

```python
# split
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100)
splits = text_splitter.split_documents(data)
```

```python
embedding = OllamaEmbeddings(model="nomic-embed-text")
vectordb = Chroma.from_documents(documents=splits, embedding=embedding)
vectordb
```

```python
# query it
query = "What does the spreaker believe is the main reason the United States enter the war?"
docs = vectordb.similarity_search(query)
docs[0].page_content
```

```python
# Saving to the disk
vectordb = Chroma.from_documents(documents=splits, embedding=embedding, persist_directory="./chroma_db")
```

```python
# load from disk
db2 = Chroma(persist_directory="./chroma_db", embedding_function=embedding)
docs= db2.similarity_search(query)
print(docs[0].page_content)
```

```python
# Similarity Search with Score
docs = vectordb.similarity_search_with_score(query)
docs
```

```python
# Retriever option
retriever = vectordb.as_retriever()
retriever.invoke(query)[0].page_content
```
