# Ollama

Ollama supports embedding models, making it possible to build retrieval augmented genration (RAG) applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings=(
    OllamaEmbeddings(model="nomic-embed-text")  # by default its uses llama2
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
        "Alpha is the first letter of Greek alphabet",
        "Beta is the second letter of Greek alphabet",
    ]
)
```

```python
r1[0]
```

```python
print(len(r1))
print(len(r1[0]))
```

```python
embeddings.embed_query("What is the second letter of Greek alphabet")
```
