# Important Concept: Measure Before Fixing

A production performance problem should not be solved by guessing.

A useful approach is:

Measure → Locate → Diagnose → Fix → Verify

For example:

Request
    ↓
API processing
    ↓
Database
    ↓
Model inference
    ↓
Response

We need to determine which component is responsible for the increased latency.

# Interview Follow-up

The API has only 2 worker processes, but production receives 100 concurrent requests.

The ML model takes 200 ms per request.

What problem could this create, and what would you consider changing to improve throughput and latency?

---

## My Answer

There can be request queuing because there are only 2 workers and 100 requests.

The 2 workers handle requests while the other requests wait in the queue.

We may need to increase the number of workers appropriately so more requests can be handled and fewer requests have to wait in the queue.

---

## Evaluation

### Score: 8/10

This answer correctly identified the main problem.

### What I Got Right

- Identified request queueing.
- Connected limited worker capacity with high traffic.
- Understood that waiting requests increase latency.
- Suggested increasing capacity appropriately.

### What I Need to Improve

Increasing workers is not always the correct solution.

For ML applications, each worker may load a model into memory.

For example:

2 workers → 2 model copies

Increasing to:

10 workers → potentially 10 model copies

could consume significant RAM or GPU memory.

Therefore, we need to understand the architecture and resource constraints before increasing workers.

# Key Concepts

## 1. Latency

Latency is the amount of time required to complete a request.

Example:

Request sent
    ↓
Processing
    ↓
Response received

If this takes 4 seconds:

Latency = 4 seconds

---

## 2. Throughput

Throughput describes how many requests a system can process during a period of time.

Example:

100 requests/second

Higher throughput generally means the system can process more work, but throughput and latency are different measurements.

---

## 3. Worker Process

A worker is a process or execution unit used by an application server to handle requests.

In a simplified model:

Worker 1 → Request
Worker 2 → Request

If additional requests arrive while available workers are busy, requests may wait.

---

## 4. Request Queueing

When incoming requests are greater than the system's immediate processing capacity, requests may wait in a queue.

Example:

100 requests
     ↓
2 available workers
     ↓
Some requests execute
     ↓
Other requests wait
     ↓
Workers become available
     ↓
Waiting requests execute

The waiting time contributes to overall latency.

---

## 5. Concurrency

Concurrency means dealing with multiple tasks that are in progress during overlapping periods.

A system receiving many simultaneous requests needs an appropriate concurrency model.

The exact behavior depends on the server architecture, such as:

- Processes
- Threads
- Async I/O
- Synchronous code
- Asynchronous code
- CPU-bound work
- I/O-bound work

---

## 6. Resource Saturation

High traffic can cause system resources to become saturated.

Important resources to monitor include:

- CPU
- RAM
- GPU
- VRAM
- Network
- Database connections
- Worker utilization

---

# Production Debugging Pattern

For performance problems, remember:

## Measure → Locate → Diagnose → Fix → Verify

### Measure

Collect:

- Latency
- Throughput
- Error rate
- CPU
- Memory
- GPU
- Queue length

### Locate

Determine where time is being spent.

### Diagnose

Find the actual bottleneck.

### Fix

Apply an appropriate solution.

Possible solutions include:

- Increasing capacity
- Horizontal scaling
- Optimizing blocking operations
- Caching
- Batching
- Database optimization
- Better concurrency
- Autoscaling

### Verify

Measure again after the change.

Never assume the problem is fixed without validating the result.

---

# Important Interview Insight

Do not immediately say:

"Increase the number of workers."

A stronger answer is:

"I would first measure worker utilization, queueing, CPU/memory/GPU usage, and model memory requirements. Based on the bottleneck, I would decide whether to increase workers, scale horizontally, optimize the request path, or change the model-serving architecture."

This demonstrates engineering judgment.

---

# Strong Interview Answer

A strong fresher-level answer could be:

"If an ML inference API becomes much slower under high traffic, I would first measure the end-to-end latency and break the request into different components using logs, metrics, and tracing.

Since the model itself takes only around 200 ms, I would investigate the API layer and surrounding components, such as request queueing, worker utilization, database or external API calls, CPU and memory usage, and connection pools.

If I discover that the API has insufficient worker capacity and requests are waiting in a queue, I would evaluate increasing the worker capacity or scaling the service horizontally.

However, I would not blindly increase workers because each worker may consume additional memory or GPU resources, especially if it loads its own copy of the ML model.

Finally, I would load-test the system under realistic traffic and compare latency, throughput, resource utilization, and error rates before and after the change."

---

# Interview Framework Learned

For production performance questions:

1. Measure
2. Find the bottleneck
3. Understand why it happens
4. Choose the appropriate solution
5. Consider trade-offs
6. Load-test
7. Verify the improvement
