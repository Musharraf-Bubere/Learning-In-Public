# Key Concepts

## 1. Latency

Latency is the amount of time required to complete a request.

Example:

Request sent
    ↓
Processing
    ↓
Response received

If this takes 4 seconds:

Latency = 4 seconds

---

## 2. Throughput

Throughput describes how many requests a system can process during a period of time.

Example:

100 requests/second

Higher throughput generally means the system can process more work, but throughput and latency are different measurements.

---

## 3. Worker Process

A worker is a process or execution unit used by an application server to handle requests.

In a simplified model:

Worker 1 → Request
Worker 2 → Request

If additional requests arrive while available workers are busy, requests may wait.

---

## 4. Request Queueing

When incoming requests are greater than the system's immediate processing capacity, requests may wait in a queue.

Example:

100 requests
     ↓
2 available workers
     ↓
Some requests execute
     ↓
Other requests wait
     ↓
Workers become available
     ↓
Waiting requests execute

The waiting time contributes to overall latency.

---

## 5. Concurrency

Concurrency means dealing with multiple tasks that are in progress during overlapping periods.

A system receiving many simultaneous requests needs an appropriate concurrency model.

The exact behavior depends on the server architecture, such as:

- Processes
- Threads
- Async I/O
- Synchronous code
- Asynchronous code
- CPU-bound work
- I/O-bound work

---

## 6. Resource Saturation

High traffic can cause system resources to become saturated.

Important resources to monitor include:

- CPU
- RAM
- GPU
- VRAM
- Network
- Database connections
- Worker utilization

---

# Production Debugging Pattern

For performance problems, remember:

## Measure → Locate → Diagnose → Fix → Verify

### Measure

Collect:

- Latency
- Throughput
- Error rate
- CPU
- Memory
- GPU
- Queue length

### Locate

Determine where time is being spent.

### Diagnose

Find the actual bottleneck.

### Fix

Apply an appropriate solution.

Possible solutions include:

- Increasing capacity
- Horizontal scaling
- Optimizing blocking operations
- Caching
- Batching
- Database optimization
- Better concurrency
- Autoscaling

### Verify

Measure again after the change.

Never assume the problem is fixed without validating the result.

---

# Important Interview Insight

Do not immediately say:

"Increase the number of workers."

A stronger answer is:

"I would first measure worker utilization, queueing, CPU/memory/GPU usage, and model memory requirements. Based on the bottleneck, I would decide whether to increase workers, scale horizontally, optimize the request path, or change the model-serving architecture."

This demonstrates engineering judgment.

---

# Strong Interview Answer

A strong fresher-level answer could be:

"If an ML inference API becomes much slower under high traffic, I would first measure the end-to-end latency and break the request into different components using logs, metrics, and tracing.

Since the model itself takes only around 200 ms, I would investigate the API layer and surrounding components, such as request queueing, worker utilization, database or external API calls, CPU and memory usage, and connection pools.

If I discover that the API has insufficient worker capacity and requests are waiting in a queue, I would evaluate increasing the worker capacity or scaling the service horizontally.

However, I would not blindly increase workers because each worker may consume additional memory or GPU resources, especially if it loads its own copy of the ML model.

Finally, I would load-test the system under realistic traffic and compare latency, throughput, resource utilization, and error rates before and after the change."

---

7. Verify the improvement

# Concept Explanation

## Features (X)

Features are the input variables given to a machine learning model.

For example, when predicting whether a student will pass, the features could be:

- Math marks
- English marks
- Attendance
- Study hours

Example:

X = Math Marks, English Marks, Attendance, Study Hours

---

## Target (Y)

The target is the output that a supervised learning model is trained to predict.

For example:

Y = Pass/Fail

Example dataset:

| Math | English | Attendance | Study Hours | Pass/Fail |
|---:|---:|---:|---:|---|
| 45 | 87 | 80% | 2 | Fail |
| 87 | 67 | 90% | 5 | Pass |
| 90 | 90 | 95% | 6 | Pass |

Here:

X = Math, English, Attendance, Study Hours

Y = Pass/Fail

The model learns the relationship between X and Y.

---

# Supervised Learning

Supervised learning is a type of machine learning where the model is trained using labeled data.

The training data contains:

- Input features (X)
- Target/output (Y)

The model learns the relationship between the input features and the target.

Example:

Student Features → ML Model → Pass/Fail

For example:

Math = 90
English = 90
Attendance = 95%

The model may predict:

Pass

---

# Types of Supervised Learning

## 1. Classification

Classification predicts a category or class.

Examples:

- Spam / Not Spam
- Pass / Fail
- Fraud / Not Fraud
- Positive / Negative
- Cat / Dog

Example:

Email → Spam or Not Spam

The output belongs to a category.

---

## 2. Regression

Regression predicts a continuous numerical value.

Examples:

- House price
- Salary
- Temperature
- Sales
- Demand

Example:

House Features → Predicted House Price

The output could be:

₹50,00,000

rather than a category such as High or Low.

---

# Unsupervised Learning

Unsupervised learning is a type of machine learning where the training data does not contain a predefined target/output label.

Important:

Unsupervised learning still has input features.

The difference is that there is no known target Y provided during training.

General idea:

Input Features → ML Algorithm → Patterns/Groups/Structure

---

# Example of Unsupervised Learning

## Customer Segmentation

Suppose we have customer data:

| Customer | Age | Income | Spending |
|---|---:|---:|---:|
| A | 22 | 30000 | 5000 |
| B | 25 | 35000 | 6000 |
| C | 45 | 100000 | 40000 |
| D | 48 | 120000 | 50000 |

We do not provide predefined labels such as:

A → Low Value

B → Low Value

C → High Value

D → High Value

Instead, a clustering algorithm can discover groups based on similarities in the data.

For example:

Group 1 → Lower spending customers

Group 2 → Higher spending customers

The groups are discovered from the data rather than being provided beforehand.

---

# Common Unsupervised Learning Tasks

## 1. Clustering

Clustering groups similar data points together.

Examples:

- Customer segmentation
- Grouping similar documents
- Grouping similar products

Common algorithm:

K-Means

---

## 2. Dimensionality Reduction

Dimensionality reduction reduces the number of features while trying to preserve useful information.

Examples:

- PCA
- t-SNE
- UMAP

Uses include:

- Visualization
- Feature reduction
- Noise reduction

# Main Difference

| Supervised Learning | Unsupervised Learning |
|---|---|
| Uses labeled training data | Uses data without target labels |
| Has input features X | Has input features X |
| Has target Y during training | Does not have target Y during training |
| Learns relationship between X and Y | Discovers patterns or structure |
| Classification and regression are common tasks | Clustering and dimensionality reduction are common tasks |
| Example: spam detection | Example: customer segmentation |

---

# Simple Way to Remember

Supervised Learning:

X + Y → Learn relationship → Predict Y

Unsupervised Learning:

X → Discover patterns/groups

Another simple memory trick:

Supervised learning is like learning with a teacher because the correct answer is provided during training.

Unsupervised learning is like finding patterns on your own because predefined answers are not provided.

# Best Fresher Interview Answer

Classification and regression are both types of supervised learning.

Classification is used when the model needs to predict a category or class. For example, an email classification model can predict whether an email is Spam or Not Spam.

Regression is used when the model needs to predict a continuous numerical value. For example, a regression model can predict the price of a house based on features such as area, location, and number of rooms.

The main difference is that classification predicts discrete categories, while regression predicts continuous numerical values.
