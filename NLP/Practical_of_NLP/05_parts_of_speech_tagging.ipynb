# Parts Of Speech (POS) Tags in NLP

POS Tagging = Identifying grammatical role of words in a sentence.

Example:
"The dog is running fast."

The     -> DT
dog     -> NN
is      -> VBZ
running -> VBG
fast    -> RB


==================================================
1. NOUNS
==================================================

NN    -> Noun, singular
         Example: dog, car

NNS   -> Noun plural
         Example: dogs, cars

NNP   -> Proper noun singular
         Example: India, John

NNPS  -> Proper noun plural
         Example: Indians, Americans


==================================================
2. PRONOUNS
==================================================

PRP   -> Personal pronoun
         Example: he, she, they

PRP$  -> Possessive pronoun
         Example: his, her, their

WP    -> Wh-pronoun
         Example: who, what

WP$   -> Possessive wh-pronoun
         Example: whose


==================================================
3. VERBS
==================================================

VB    -> Base verb
         Example: eat, go

VBD   -> Past tense verb
         Example: ate, went

VBG   -> Verb gerund/present participle
         Example: eating, going

VBN   -> Past participle
         Example: eaten, gone

VBP   -> Verb present tense
         Example: eat

VBZ   -> Verb 3rd person singular
         Example: eats


==================================================
4. ADJECTIVES
==================================================

JJ    -> Adjective
         Example: beautiful

JJR   -> Comparative adjective
         Example: better

JJS   -> Superlative adjective
         Example: best


==================================================
5. ADVERBS
==================================================

RB    -> Adverb
         Example: quickly

RBR   -> Comparative adverb
         Example: faster

RBS   -> Superlative adverb
         Example: fastest

WRB   -> Wh-adverb
         Example: when, where


==================================================
6. DETERMINERS & ARTICLES
==================================================

DT    -> Determiner
         Example: the, a

PDT   -> Predeterminer
         Example: all, both

WDT   -> Wh-determiner
         Example: which


==================================================
7. CONJUNCTIONS
==================================================

CC    -> Coordinating conjunction
         Example: and, but


==================================================
8. PREPOSITIONS
==================================================

IN    -> Preposition/Subordinating conjunction
         Example: in, on, because


==================================================
9. NUMBERS
==================================================

CD    -> Cardinal number
         Example: one, 100


==================================================
10. MODAL VERBS
==================================================

MD    -> Modal
         Example: can, may, should


==================================================
11. PARTICLES
==================================================

RP    -> Particle
         Example: up, off

Example:
give up
take off


==================================================
12. TO
==================================================

TO    -> to
         Example: to go


==================================================
13. INTERJECTIONS
==================================================

UH    -> Interjection
         Example: wow, oh


==================================================
14. SPECIAL TAGS
==================================================

EX    -> Existential there
FW    -> Foreign word
LS    -> List item marker
POS   -> Possessive ending
SYM   -> Symbol
.     -> Sentence ending punctuation
,     -> Comma
:     -> Colon
(     -> Left bracket
)     -> Right bracket
``    -> Opening quotation
''    -> Closing quotation


==================================================
MOST IMPORTANT TAGS FOR INTERVIEW
==================================================

NN    -> Noun
NNS   -> Plural noun
VB    -> Verb
VBD   -> Past tense verb
VBG   -> Gerund
JJ    -> Adjective
RB    -> Adverb
PRP   -> Pronoun
DT    -> Determiner
IN    -> Preposition
CC    -> Conjunction


==================================================
REAL NLP EXAMPLE
==================================================

Sentence:
"John is eating pizza quickly."

John     -> NNP
is       -> VBZ
eating   -> VBG
pizza    -> NN
quickly  -> RB


==================================================
WHY POS TAGGING IS IMPORTANT
==================================================

- Sentence understanding
- Chatbots
- Machine Translation
- Text Summarization
- Named Entity Recognition (NER)
- Speech Recognition
- Grammar Checking

NLTK POS TAGGING CODE
==================================================

import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

sentence = "John is eating pizza quickly"

words = word_tokenize(sentence)

print(pos_tag(words))


Output:

[('John', 'NNP'),
 ('is', 'VBZ'),
 ('eating', 'VBG'),
 ('pizza', 'NN'),
 ('quickly', 'RB')]

```python
# Speech Of DR APJ Abdul Kalam
paragraph = """I have three visions for India. In 3000 years of our history, people from all over 
               the world have come and invaded us, captured our lands, conquered our minds. 
               From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
               the French, the Dutch, all of them came and looted us, took over what was ours. 
               Yet we have not done this to any other nation. We have not conquered anyone. 
               We have not grabbed their land, their culture, 
               their history and tried to enforce our way of life on them. 
               Why? Because we respect the freedom of others.That is why my 
               first vision is that of freedom. I believe that India got its first vision of 
               this in 1857, when we started the War of Independence. It is this freedom that
               we must protect and nurture and build on. If we are not free, no one will respect us.
               My second vision for India’s development. For fifty years we have been a developing nation.
               It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
               in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
               Our achievements are being globally recognised today. Yet we lack the self-confidence to
               see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
               I have a third vision. India must stand up to the world. Because I believe that unless India 
               stands up to the world, no one will respect us. Only strength respects strength. We must be 
               strong not only as a military power but also as an economic power. Both must go hand-in-hand. 
               My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of 
               space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
               I was lucky to have worked with all three of them closely and consider this the great opportunity of my life. 
               I see four milestones in my career"""
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
stemmer = PorterStemmer()
sentences = sent_tokenize(paragraph)
```

```python
sentences
```

```python
# We will find the POS Tag

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [word for word in words if word not in set(stopwords.words('english'))]
    # sentences[i] = ' '.join(words) # converting all the list of words into sentences
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
```
